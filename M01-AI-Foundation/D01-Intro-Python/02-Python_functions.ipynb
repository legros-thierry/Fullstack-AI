## Functions 🐍

A **function** is a reusable block of code that does one specific task. You define it once with `def`. Then you call it whenever you need it. You can call it once, a thousand times, or a million times. **You can call it from different places.** You can even call it from inside other functions.

<img src="https://ai-essentials-assets.s3.eu-west-3.amazonaws.com/M01-Python_for_AI/fsai-M01-D01-What_is_a_function.png" />

**Parameters** are the inputs of the function. The `return` statement sends the output back to the caller.

Here is how to define a function:

```python
def function_name(parameter1, parameter2):
    # code to run
    return result
```

Here is how to call a function:

```python
result = function_name(argument1, argument2)
```

### Your first function

In the previous lecture, you wrote an `if/elif/else` block that labels a drink as `"high"`, `"medium"`, or `"low"` based on its cup count. Every time you wanted to reuse that logic, you had to rewrite it. Let's turn it into a function so you can call it with a single line.

The next code block defines a function called `classify_popularity`. It takes one input and returns one output.

````python
# Define a function to sort drink popularity
def classify_popularity(cups_sold):
    """Sort drink popularity based on cups sold.

    Parameters:
        cups_sold: integer count of cups sold

    Returns:
        string: "high", "medium", or "low"
    """
    if cups_sold >= 100:
        return "high"
    elif cups_sold >= 50:
        return "medium"
    else:
        return "low"
````

````python
# Call the function with different cups-sold counts
print(classify_popularity(142))   # high
print(classify_popularity(67))    # medium
print(classify_popularity(34))    # low
````

`def classify_popularity(cups_sold):` creates a function named `classify_popularity`. It takes one input called `cups_sold`. Inside the function, the `if/elif/else` block returns a different string based on the value. `return` sends the result back to whoever called the function.

The triple-quoted text right after `def` is a **docstring** (a string that describes what the function does). Professional Python code always has docstrings. They are the first thing a new teammate reads. They are also what Python shows when you call `help(classify_popularity)`.

`classify_popularity(142)` runs the function with the argument `142`. Inside, `cups_sold` holds the value `142`. The first `if` matches, so the function returns `"high"`. `print()` then displays that result.

One function call replaces five lines of conditional logic. The logic lives in one place. If you change the thresholds tomorrow, you change them once.

### Functions that call other functions

Functions can call other functions. This is how you build complex systems from simple pieces. Each function stays small. Each function does one thing. You connect them to build something bigger.

The next code block defines `format_order_log`. This function calls `classify_popularity` inside it to get the popularity level. Then it builds a full log entry from several pieces.

````python
def format_order_log(drink, cups_sold, loyalty_member):
    """Make a formatted log entry for pipeline output."""
    popularity = classify_popularity(cups_sold)
    popularity_upper = popularity.upper()

    if loyalty_member:
        loyalty_tag = "LOYAL"
    else:
        loyalty_tag = "GUEST"

    log_entry = f"[{popularity_upper}] [{loyalty_tag}] {drink} ({cups_sold} cups)"
    return log_entry

result = format_order_log(
    "Oat Milk Caramel Latte",
    142,
    True
)
print(result)
````

`format_order_log` takes three inputs: `drink`, `cups_sold`, and `loyalty_member`. It calls `classify_popularity(cups_sold)` to get the popularity level. Then it turns the popularity string into uppercase with `.upper()`. The `if/else` block picks the loyalty tag. Finally, the f-string builds the full log entry and `return` sends it back.

This pattern of building small functions and joining them is the base of maintainable code. Each function does one thing. You can test and fix each one on its own. If the format of the log entry changes, you only touch `format_order_log`. The classification logic stays safe in `classify_popularity`.

### Default parameter values

Some parameters have sensible defaults. Callers can skip them when the default is fine. You set a default value with `=` in the function definition. This makes function calls shorter. It also makes the common case easy and the special case still possible.

The next code block shows a search function with optional parameters. Only the category is required. The other two parameters have defaults.

````python
def search_drinks(category, min_cups=0, loyalty_only=False):
    """Mimic a BeanStream menu search."""
    query = f"Searching for '{category}'"

    if min_cups > 0:
        query = query + f" with at least {min_cups} cups sold"

    if loyalty_only:
        query = query + " (loyalty customers only)"

    print(query)


# Call with only the required argument
search_drinks("Cold Brew")

# Change one default
search_drinks("Cold Brew", min_cups=50)

# Change both defaults
search_drinks("Cold Brew", min_cups=50, loyalty_only=True)
````

`min_cups=0` and `loyalty_only=False` are default values. If you do not pass them when calling the function, Python uses the defaults. The first call uses only `"Cold Brew"`. The second call changes `min_cups` to 50. The third call changes both.

Using named arguments like `min_cups=50` makes the code clear and easy to read. Anyone looking at `search_drinks("Cold Brew", min_cups=50, loyalty_only=True)` understands what each value means. This is much better than `search_drinks("Cold Brew", 50, True)`, where you would have to check the function definition to know what `50` and `True` mean.

<Note type="tip" title="Order matters in function definitions">

Required parameters must come before optional ones. `def f(a, b=0)` works. `def f(a=0, b)` does not. Put the parameters a caller *must* provide first, and the parameters a caller *may* provide last.

</Note>

## Error handling

When Python hits a problem it cannot solve, it **raises an exception** (an error object) and stops the program. You have seen this before. Divide by zero. Convert `"N/A"` to an integer. Read a missing dictionary key. Each of these raises a specific exception type and halts everything.

In a small script, a crash is fine. You read the error, fix the code, and rerun. But in a data pipeline, one bad record out of ten thousand should not stop everything. You need a way to say: "try this, and if it fails, do this other thing instead."

Python gives you the `try/except` pattern for exactly this. Code inside `try` runs normally. If any line raises an exception, Python jumps to the matching `except` block and runs that instead. The program keeps going.

To make debugging easier, `try/except` is often used with specific error types like `ValueError`, `TypeError`, `KeyError`, and others. These **[built-in exceptions](https://docs.python.org/3/library/exceptions.html)** cover the most common problems. You can handle each one differently.

### A basic try/except

The next code block shows what happens without and with error handling. Without the `try/except`, the line `int("not_a_number")` crashes the whole program. With it, Python catches the error and keeps going.

````python
# Without error handling, this line would crash:
# result = int("not_a_number")  # ValueError!

# With error handling: safe recovery
try:
    result = int("not_a_number")
    print(f"Converted: {result}")
except ValueError:
    print("Could not convert. Skipping this value.")

print("Pipeline keeps running...")
````

Python tries to run the code inside the `try` block. `int("not_a_number")` fails because `"not_a_number"` is not a valid number. Python raises a `ValueError`. Instead of crashing, the `except ValueError` block runs. It prints a warning message. Then the program keeps running from the next line.

The final line `print("Pipeline keeps running...")` always runs, no matter what happened in the `try` block. That is the whole point of `try/except`. Your pipeline survives bad data.

### Wrapping error handling into a reusable function

The previous block caught one error in one place. But you will convert strings to integers dozens of times in a real pipeline. It would be painful to write `try/except` every single time. The better approach is to wrap the pattern in a function.

The next code block defines `safe_int`. It tries to convert any value to an integer. If the conversion fails, it returns a default value instead of crashing. You can now call `safe_int(x)` anywhere in your pipeline and trust that it will never crash.

````python
def safe_int(value, default=0):
    """Safely convert a value to integer.

    Returns default if conversion fails.
    """
    try:
        return int(value)
    except (ValueError, TypeError):
        return default


print(safe_int("142"))     # 142
print(safe_int("N/A"))     # 0
print(safe_int(None))      # 0
print(safe_int(""))        # 0
````

`safe_int` takes two parameters. `value` is the input to convert. `default` has a default value of `0`. Inside, the `try` block calls `int(value)`. If the value is a valid number string like `"142"`, the function returns `142`. If the value is bad text like `"N/A"`, an empty string, or `None`, Python raises an exception.

The `except (ValueError, TypeError)` block catches two exception types at once. `ValueError` happens when the text cannot be parsed as a number. `TypeError` happens when the input is not a text or number at all, like `None`. Catching both keeps the function robust against several kinds of bad input.

One function now handles four different kinds of bad input without crashing. This is real production code. A pipeline that reads millions of records will hit every single one of these cases.

<Note type="important" title="Catch specific exception types">

Always catch **specific** exception types. A bare `except:` catches everything, including logic bugs you need to see. Catch `ValueError` and `TypeError` when converting data. Catch `KeyError` when reading dictionaries. Let unexpected exceptions come up so you can fix the real cause.

</Note>

## Error handling in batch processing

Now, imagine you get data from the [BeanStream ordering API](https://api.beanstream.com/). You want to process a batch of orders. But some records have bad fields. Missing drinks. Cup counts with text in them. Invalid loyalty flags.

You do not want one bad record to crash the whole pipeline. You want to log the problem, skip the bad record, and keep processing. This is where functions and error handling come together. You wrap one record's logic in a function with a `try/except`. Then you loop through the batch and call that function for every record.

The next code block shows this exact pattern. `process_order` handles one record safely. The main loop calls it for every record in the batch.

````python
def process_order(raw):
    """Process one raw order record safely.

    Returns a cleaned dictionary, or None if the record is bad.
    """
    try:
        drink = raw.get("drink", "Unknown")
        cups_raw = raw.get("cups_sold", 0)
        cups_int = int(cups_raw)
        loyalty_flag = raw.get("loyalty_member", False)

        result = {
            "drink": drink,
            "cups_sold": cups_int,
            "loyalty_member": loyalty_flag
        }
        return result

    except (ValueError, TypeError) as e:
        print(f"  WARNING: Skipping bad record. {e}")
        return None
````

````python
# Data received with one bad record
raw_batch = [
    {"drink": "Good Latte", "cups_sold": "142", "loyalty_member": True},
    {"drink": "Broken Drink", "cups_sold": "N/A", "loyalty_member": False},
    {"drink": "Cold Brew", "cups_sold": "201", "loyalty_member": True}
]

# Process every record, skip the bad ones
cleaned_batch = []
for raw in raw_batch:
    cleaned = process_order(raw)
    if cleaned is not None:
        cleaned_batch.append(cleaned)

print(f"\nProcessed {len(cleaned_batch)} of {len(raw_batch)} records.")
for order in cleaned_batch:
    print(f"  {order}")
````

Let's walk through this piece by piece.

`process_order` tries to read each field from the raw dictionary. It uses `.get()` for safe access. `.get("drink", "Unknown")` returns `"Unknown"` if the key is missing, instead of crashing. This handles missing fields.

`int(cups_raw)` tries to convert the cups count from a string to an integer. If the raw value is `"N/A"`, this fails with a `ValueError`. The `except` block catches it. The function prints a warning and returns `None`.

The main loop calls `process_order` for every record in the batch. It checks if the result is `None`. If not, it appends the cleaned record to `cleaned_batch`. At the end, 2 out of 3 orders were processed. The bad record was logged but did not stop the pipeline.

This is the shape of every production data pipeline. One function per record. Error handling inside the function. A loop that calls it for every record. Bad data is logged and skipped. The pipeline always finishes.
