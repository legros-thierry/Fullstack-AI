## What is a code editor?

Code editors care about **syntax**: the structure and rules of a programming language. They color different parts of your code so you can see its structure. They mark mistakes before you even run the code. They auto-complete names so you type less and make fewer typos.

<img src="https://ai-fullstack-assets.s3.eu-west-3.amazonaws.com/M01-AI_Engineering_Foundation/AIFS-M01-D01-IDE.png"/>

You could write Python in Notepad or TextEdit. But you would have no colors, no error warnings, no auto-complete, and no way to run your code from the same window. That is like making a latte with only a kettle and a spoon when a full espresso machine is right there.

## Why Cursor?

**Cursor** is a code editor built on top of VS Code (Microsoft's popular editor). It adds something no old-style editor offers: an **AI assistant built right into your workflow**.

In a normal editor, you write code, run it, see an error, search the internet for help, read a Stack Overflow answer, or use ChatGPT, then come back to the editor and try again. In Cursor, you write code, see an error, and ask the built-in AI to explain the error and suggest a fix. The AI reads your code, understands the context, and answers inside the editor. No tab switching. No copy-pasting error messages.

<Note type="tip" title="AI help is not a shortcut">

Cursor's AI helps you write code faster. But it does not replace understanding. When the AI suggests a solution, read it. Understand why it works. If you accept suggestions blindly without learning, you build weak skills. The goal is to use AI as a **thinking partner**, not a replacement for thinking.

For those of you who work with other IDEs (Integrated Development Environments, which means full-feature code editors), the ideas in this notebook still apply. GitHub Copilot and Tabnine are two good choices.

## Setting up Python

**Python** is the programming language you use through this whole curriculum. It is a separate program that lives on your computer. Cursor is where you *write* Python code. Python itself is the engine that *runs* it.

Think of it this way. Cursor is the notebook you write your coffee recipes in. Python is the espresso machine that actually makes the coffee. You need both.

### Step 1: Check if Python is already installed

Open the **terminal** inside Cursor.

<img src ="https://ai-fullstack-assets.s3.eu-west-3.amazonaws.com/M01-AI_Engineering_Foundation/AIE-M01-D01-Cursor_terminal.png" />

This is a **command line interface** (a window where you type text commands, and the computer runs them).

Type the following command and press Enter:

```bash
python --version
```

If Python is installed, you see something like: `Python 3.12.4`. The exact numbers do not matter, as long as the first number is 3 (meaning Python 3, not the old Python 2).

If the command fails or shows Python 2.x, you need to install Python 3.

### Step 2: Install Python (if needed)

<img src="https://ai-fullstack-assets.s3.eu-west-3.amazonaws.com/M01-AI_Engineering_Foundation/AIE-M01-D01-Python_install.png" />

**On macOS:** The simplest way is to install Python from [python.org](https://www.python.org/downloads/). Download the latest Python 3 release. Run the installer. Check the box "Add Python to PATH" during installation.

**On Windows:** Download the latest Python 3 from [python.org](https://www.python.org/downloads/). Run the installer. **Very important:** check the box **"Add Python to PATH"** on the first screen. This lets Cursor's terminal find Python. Without this checkbox, you have to set the path by hand.

**On Linux:** Most distributions include Python 3. If not, use your package manager: `sudo apt install python3` on Ubuntu or Debian.

### Step 3: Verify in Cursor

After installation, close and reopen Cursor's terminal. Type `python --version` again. You should see Python 3.x.x. If you do, congratulations! Your Python environment is ready. You can now write and run Python code right from Cursor.

### Step 4 (Recommended): Install Python extension in Cursor

Cursor is built on VS Code, which has a rich ecosystem of extensions. The **Python extension** adds features like syntax highlighting, error checking, and code completion for Python.

To install it, go to the Extensions view in Cursor (click the square icon on the left sidebar), search for "Python", and install the extension published by Microsoft. This will improve your Python coding experience in Cursor.

### Step 5 (Recommended): Install Jupyter extension in Cursor

The **Jupyter extension** lets you create and run Jupyter notebooks right in Cursor. Notebooks are interactive documents that mix code, text, and charts. They are used a lot in data science and AI for trying things out and sharing results.

To install it, go to the Extensions view in Cursor, search for "Jupyter", and install the extension published by Microsoft. This will let you create `.ipynb` files and run code cells one by one inside Cursor.

## The terminal

The terminal (also called **command line** or **shell**) is a text-based window on your computer. Instead of clicking icons, you type commands. This feels strange if you have only used graphical interfaces. But engineers use the terminal all the time because it is faster and more exact than clicking through menus.

<Video video="https://vimeo.com/1185930044/a061a8590b?share=copy&fl=sv&fe=ci" />

<Note type="tip" title="The terminal is your friend">

If you are on Mac or Linux, the terminal is your best friend. If you are on Windows, the terminal is still key, but you may also use PowerShell or Command Prompt. The goal is to have a reliable way to run commands and manage your files without leaving the keyboard.

Here are some basic terminal commands for Mac or Linux users that are useful to know:

| Command | What it does | Example |
|---------|-------------|---------|
| `ls` | Lists files in the current directory | `ls` |
| `pwd` | Shows the current directory path | `pwd` |
| `cd folder` | Changes directory to "folder" | `cd Documents` |
| `mkdir folder` | Creates a new folder named "folder" | `mkdir my_project` |
| `rm file` | Deletes a file named "file" | `rm old_script.py` |    
| `rm -r folder` | Deletes a folder and its contents | `rm -r old_project` |
| `clear` | Clears the terminal screen | `clear` |
| `mv` | Moves or renames a file or folder | `mv old_name new_name` |
| `echo` | Prints text to the terminal | `echo Hello, world!` |
| `>`| Redirects output to a file | `echo Hello > greeting.txt` |

</Note>

**In Cursor, the terminal lives at the bottom of the window.** You will use it to run Python files, install libraries, and manage your project files.

Here are the three commands you need today:

| Command | What it does | Example |
|---------|-------------|---------|
| `python --version` | Shows which Python version is installed | `Python 3.12.4` |
| `python filename.py` | Runs a Python file | `python hello.py` |
| `pip install package` | Installs a Python library | `pip install pandas` |

You do not need to memorize more commands right now. These three carry you through the whole day.

## Notebooks versus scripts

<Video video = "https://vimeo.com/1185930037/74a880ea91?share=copy&fl=sv&fe=ci" />

You are reading this content inside a **Jupyter Notebook** (a `.ipynb` file). Notebooks mix text explanations with runnable code cells. They are great for learning, trying things out, and showing analysis.

In production (real live systems), your pipeline code lives in **Python scripts** (`.py` files). Scripts have only code and comments. They run from start to finish without stops. Cursor is great at editing both notebooks and scripts.

| Feature | Notebook (`.ipynb`) | Script (`.py`) |
|---------|-------------------|----------------|
| Run cells one by one | Yes | No. Runs top to bottom |
| Mix text and code | Yes | Code only (with comments) |
| Best for | Learning, exploring, reports | Production pipelines, automation |
| Used in | Jupyter, Cursor, Colab | Terminal, CI/CD, cron jobs |

## Cursor's AI features in practice

Cursor's AI features set it apart from every other editor. Here are the three features you will use most.

### Inline chat (Ctrl+K / Cmd+K)

Select a piece of code. Press **Ctrl+K** (Windows/Linux) or **Cmd+K** (macOS). A chat box appears right there. Type a question or command: "explain this code", "add error handling", "rename variables to snake_case". Cursor's AI reads the selected code and answers right in the editor.

### Side chat (Ctrl+L / Cmd+L)

Press **Ctrl+L** (Windows/Linux) or **Cmd+L** (macOS) to open a conversation panel on the side. This panel works like a chatbot that knows your whole project. Ask it to explain concepts, debug errors, or write code. The AI sees your open files and can talk about them in its answers.

### Tab completion

As you type, Cursor suggests completions. Press **Tab** to accept. These suggestions go beyond simple auto-complete. The AI guesses what you want to write based on the context. Your current function, the code around it, and even the variable names you have used. Accepting a suggestion is optional. Always review what the AI suggests before you accept it.

<Note type="important" title="Use AI features actively">

Do not wait until you are stuck to use Cursor's AI. Ask it to explain code you wrote but do not fully understand. Ask it to suggest improvements. Ask it what a specific error message means. The more you talk with the AI, the faster you learn. Think of it as having a patient senior engineer sitting next to you, always ready to help.

</Note>
