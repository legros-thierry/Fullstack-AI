
A customer asks for a topping stored in the variable `requested_topping`. Check if it belongs to the menu. If yes, print `"Adding [topping] to the pizza."`. If no, print `"Sorry, we don't have [topping]."`.

Use the `in` operator to check if the topping belongs to the list. Test with these values: `"mushroom"`, `"pineapple"`, and `"ham"`.

````python
menu_toppings = ["mushroom", "pepperoni", "olive", "onion", "pepper", "ham"]
requested_topping = "banane"
````

````python
def check_topping(menu_toppings, requested_topping):
    isOK = False
    if requested_topping not in menu_toppings:
        print(f"Sorry, we don't have {requested_topping}.")
        isOk=False
    else:
        print(f"Adding {requested_topping} to the pizza.") 
        isOK=True
    return isOK
````
