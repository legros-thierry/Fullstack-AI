## Task 1: Bake 5 pizzas in a row 

Mario has 5 pizzas waiting. He must put each one in the oven and print `"Pizza N is baking..."` where N is the number of the pizza.

Use a `for` loop with `range()` to count from 1 to 5. Inside the loop, print the baking message.

<Note type="hint">
`range(1, 6)` gives you numbers 1 to 5. The second number is not included. So you need `range(1, 6)` to reach 5.
</Note>

````python
# YOUR CODE HERE
for pizza in range(1, 6):
    print(f"Pizza {pizza} is baking...")
````

## Task 3: Count the cheese pizzas 

Mario wants to know how many pizzas in tonight's orders use cheese. Go through this list:

```python
orders = ["margherita", "marinara", "four cheese", "vegan", "mozzarella special", "marinara"]
```

The pizza `"marinara"` and `"vegan"` have **no cheese**. All others do.

Use a `for` loop and a counter variable. For each pizza, check if it is not in the no-cheese list. If it uses cheese, add 1 to your counter.

Print the total number of cheese pizzas at the end.

<Note type="hint">
Start with `cheese_count = 0`. Inside the loop, update it with `cheese_count = cheese_count + 1`.
</Note>

````python
orders = ["margherita", "marinara", "four cheese", "vegan", "mozzarella special", "marinara"]
no_cheese = ["marinara", "vegan"]

# YOUR CODE HERE
counter = 0
for order in orders:
    if order not in no_cheese:
        counter += 1
print(f"Number of cheese pizzas: {counter}")
````

## Task 4: Skip the burnt pizzas 

Sometimes a pizza burns in the oven. Mario must throw these away and move to the next one.

You have this list of pizzas and their status:

```python
pizza_numbers = [1, 2, 3, 4, 5, 6]
burnt_pizzas = [2, 5]
```

Loop through the pizza numbers. If a pizza is burnt, print `"Pizza N burnt. Skipping."` and move on. Use `continue`. If a pizza is fine, print `"Pizza N delivered."`.

<Note type="hint">
`continue` sends the code to the next loop turn right away. The lines below `continue` do not run for that turn.
</Note>

````python
pizza_numbers = [1, 2, 3, 4, 5, 6]
burnt_pizzas = [2, 5]

# YOUR CODE HERE
for pizza in pizza_numbers:
    if pizza in burnt_pizzas:
        print(f"Pizza {pizza} is burnt. Skipping")
        continue
    print(f"Pizza {pizza} delivered.")
````

## Task 5: Find the special order 

A VIP customer placed order number 307. Mario must find this order in the queue and stop looking once he finds it.

```python
order_queue = [301, 302, 303, 304, 305, 306, 307, 308, 309]

```

Use a `for` loop to go through the queue. For each order, print `"Checking order N..."`. When you find the target, print `"🎯 Found order 307!"` and use `break` to stop.

<Note type="hint">
`break` ends the loop right away. Use it as soon as you find the order.
</Note>

````python
order_queue = [301, 302, 303, 304, 305, 306, 307, 308, 309]
target = 307

nb_test = 0
# YOUR CODE HERE
for order in order_queue:
    nb_test += 1
    print(f"Checking order {order}...")
    if order == target:
        print(f"🎯 Found order {order} !")
        break

print(f"Nombre de tests effectués: {nb_test}")
````

## Task 6: The oven time limit 

Each pizza takes 8 minutes to bake. Mario's oven runs for 40 minutes total before he must let it cool. How many pizzas can he bake before he hits the limit?

Use a `while` loop. Start with `total_time = 0` and `pizza_count = 0`. Add 8 minutes each turn. Stop when adding the next pizza would pass 40 minutes.

Print the time after each pizza. At the end, print the total pizzas baked.

<Note type="hint">
Check if `total_time + 8` is still within the limit **before** you add it. This way, you stop at the right moment.
</Note>

````python
time_limit = 40
time_per_pizza = 8
total_time = 0
pizza_count = 0

# YOUR CODE HERE
while total_time + time_per_pizza < time_limit:
    total_time += time_per_pizza
    pizza_count += 1
    print(f"the time: {total_time} min")
print(f"total de pizza cuites: {pizza_count}")
````

## Task 7: Your first function 

Mario keeps asking you: "What is the price of this pizza?" You do not want to calculate it by hand every time. You build a function.

Write a function called `pizza_price`. It takes one argument: `size`. It returns the price:

- Small: 8 euros
- Medium: 11 euros
- Large: 14 euros
- Anything else: 0

Call your function three times. Print the price for Small, Medium, and Large.

<Note type="hint">
A function starts with `def`. Inside, use `if`, `elif`, `else` to choose the price. Use `return` to send the price back.

```python
def pizza_price(size):
    if size == "Small":
        return 8
    # continue...
```
</Note>

````python
# YOUR CODE HERE
def pizza_price(size):
    price = 0
    if size.title() == "Small":
        price = 8
    elif size.title() == "Medium":
        price = 11
    elif size.title() == "Large":
        price = 14
    else:
        price = 0
    return price

print(pizza_price("Small"))
print(pizza_price("Medium"))
print(pizza_price("Large"))
````

## Task 8: A function with two inputs 

Mario wants a function that calculates the total for a full order. He sells pizzas and drinks.

Write a function called `order_total`. It takes two arguments: `pizza_count` and `drink_count`. Each pizza costs 11 euros. Each drink costs 3 euros. The function returns the total price.

Call the function three times:
- 2 pizzas, 1 drink
- 3 pizzas, 4 drinks
- 1 pizza, 0 drinks

Print the total for each call.

````python
# YOUR CODE HERE
def order_total(pizza_count, drink_count):
    total = (pizza_count * 11) + (drink_count * 3)
    return total

print(order_total(2,1))
print(order_total(3,4))
print(order_total(1,0))
````
