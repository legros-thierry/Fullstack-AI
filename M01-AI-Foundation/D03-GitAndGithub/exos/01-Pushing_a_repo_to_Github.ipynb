# Push your first repository to GitHub 🚀

Two weeks into your job at **Helix Analytics**, your team lead asks you to share a small interest calculator you wrote last week. Another analyst needs the same logic for a client report and does not want to rebuild it from scratch. Sending the notebook by email is not an option. The company stores every reusable script on GitHub so the team can find it, review it, and track changes over time.

In this exercise, you create a new GitHub repository, clone it to your machine, add your notebook, and push your first commits to the remote. By the end, your code is online and your teammate can pull it with one command.

## Task 1: Create a new repository on GitHub

Go to https://github.com/new. You will see a form that asks for a repository name, a description, and a few configuration choices.

Pick a name that describes the content. A name like `interest-calculator` tells your teammates what to expect. A name like `project1` or `test` forces them to open the repository to find out.

Set the visibility to **Private** for now. The notebook contains logic you may want to refine before showing it to a wider audience. You can switch the repository to public later from the settings page once you are happy with the code.

Tick the box **Add a README file**. This creates a `README.md` file at the root of the repository. GitHub displays this file on the repository home page, and your team uses it to document what the project does, how to install it, and which dataset it relies on. When you write that content, keep the heading hierarchy clean (a single `#` title, then `##` sections, no skipped levels) and give every image a descriptive alt text so the README stays readable with a screen reader.

In the **Add .gitignore** dropdown, select **Python**. This adds a hidden file that tells Git to skip noise like Jupyter checkpoint folders and Python cache directories. Without it, every commit ships files that no one needs to read.

Click **Create repository**.

## Task 2: Clone the repository to your machine

GitHub now shows the home page of your empty repository. Click the green **Code** button and copy the HTTPS URL. The URL ends with `.git`.

Open a terminal and navigate to the folder where you keep your work projects. Use `pwd` to see your current location, `ls` to list the folders, and `cd folder_name` to move into one of them.

Once you are in the right place, run:

```bash
git clone https://github.com/your-username/your-repo.git
```

Replace the URL with the one you copied from GitHub. Git downloads the repository into a new folder named after your repo.

Move into that folder with `cd` and confirm that you see `README.md` listed when you run `ls`. Every command from now on must run inside this folder.

## Task 3: Check the initial state of the repository

Before you add any files, ask Git to describe the current state of the repository. This habit protects you from committing the wrong files later.

Run:

```bash
git status
```

Git replies with `nothing to commit, working tree clean`. This means the local copy matches the remote copy on GitHub exactly. Nothing has changed.

## Task 4: Add your notebook to the repository

Download [Calculate_interests.ipynb](https://ai-fullstack-assets.s3.eu-west-3.amazonaws.com/M01-AI_Engineering_Foundation/Calculate_interests.ipynb) notebook and place it into the repository folder. You can drag it from a file explorer or use a terminal command like `cp` on Mac and Linux or `copy` on Windows.

If you do not have the notebook on hand, download a sample copy from the Resources section of the course and place it in the repository folder.

Run `git status` again. This time, Git lists the notebook in red under **Untracked files**. Red means Git can see the file but is not yet watching it for changes. You need to tell Git to start tracking it.

Run:

```bash
git add Calculate_interests.ipynb
```

Run `git status` one more time. The notebook now appears in green under **Changes to be committed**. The file is staged, which means Git is ready to record a snapshot of it on your next commit.

## Task 5: Record your first commit

A commit is a saved snapshot of your work with a short message that explains what changed. Your teammates read these messages later to understand the history of the project. Vague messages like `update` or `fix` waste their time.

Run:

```bash
git commit -m "Add interest calculator notebook"
```

Git confirms the commit and shows a short summary with the file count and the commit hash. The snapshot is now stored locally, but it is not yet on GitHub.

## Task 6: Make a second commit

Real projects evolve through many small commits. Practice the cycle one more time so it becomes routine.

Open the notebook in Jupyter and make a small improvement. You can add a markdown cell with a short title at the top, fix a typo, or add a comment to one of the code cells. Save the notebook.

Run `git status`. The notebook appears under **Changes not staged for commit** because Git sees that the file changed since the last snapshot.

Stage the change and commit it with a message that describes what you did:

```bash
git add Calculate_interests.ipynb
git commit -m "Add section title and inline comments"
```

Run `git log` to see the full history of commits in this repository. You should see two entries listed from newest to oldest, each with the author, the date, and the message you wrote.

## Task 7: Push your commits to GitHub

Your two commits live on your laptop. Your teammate cannot see them yet. The `git push` command uploads your local commits to the remote repository on GitHub.

Run:

```bash
git push
```

Git asks for your GitHub credentials the first time you push from a new machine. Once it finishes, open the repository page on GitHub in your browser and refresh it. Your notebook now appears in the file list at the root of the repository.

Click on the **Commits** link near the top of the page. You should see the two commits you created, with the messages you wrote and your name as the author. This is the same history you saw locally with `git log`, now visible to anyone on your team who has access to the repository.

## Task 8: Confirm the workflow with a teammate

Your team lead asks you to verify that the repository works for someone other than you. Send the repository URL to a colleague or use a second machine of your own. Ask them to run:

```bash
git clone https://github.com/your-username/your-repo.git
```

Check that the clone command downloads the notebook and the README without errors. If your colleague reports a permission error, the repository visibility is still set to private and they need an invitation to access it. Open the repository settings on GitHub, go to the **Collaborators** section, and add their GitHub username.
