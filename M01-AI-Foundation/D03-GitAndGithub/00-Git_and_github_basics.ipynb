# Git and GitHub basics 🔧

## What you will learn in this course 🧐🧐

You already know how to write code. But code alone is not enough in a real company. Your team needs to share code, track every change, and work on the same project at the same time. Without a system to manage this, projects break fast. Files get lost. Two people overwrite each other's work. Nobody knows which version is the latest. This is why every modern tech team uses **Git** (a version control tool that saves a full history of your code) and **GitHub** (a cloud platform that stores Git projects online and lets people collaborate).

Whether you work as a data analyst, a data engineer, an AI engineer, or a software developer, you will use these tools every single day. Recruiters also look at your GitHub profile before they interview you. A clean public profile with real projects is now a standard part of a tech CV.

By the end of this course, you will be able to:

- Explain what a Version Control System (VCS) is and why teams need one
- Configure Git on your computer with your username and email
- Use the core Git commands to clone, track, save, and push code changes
- Create a remote repository on GitHub and connect it to your local work
- Apply a simple team workflow with fork and pull request to contribute to a shared project
- Decide when to make a repository public or private based on the business situation

<Note type="note" title="Terminal commands you should already know">

Before you start, you should be comfortable with basic Bash commands in your terminal: `pwd`, `cd`, and `ls`. You should also know these file commands:

- `mkdir dirname` creates a new directory.
- `touch filename` creates a new empty file.
- `open filename` opens the file with the default software for its type.
- `cp existing_file new_file` copies a file. For example, `cp ../file1.ipynb folder1/file2.ipynb` copies `file1.ipynb` from the parent directory into `folder1/` and renames it `file2.ipynb`.
- `mv existing_file new_location` moves or renames a file. For example, `mv file1.md README.md` renames `file1.md` to `README.md`.

All these commands run inside your *working directory*. Use `pwd` at any time to check where you are. Use `cd` to change directory.

</Note>

## The version control problem

<img src="https://ai-fullstack-assets.s3.eu-west-3.amazonaws.com/M01-AI_Engineering_Foundation/AIFS-M01-D03-Git.png" />

Imagine you start writing the FreshCart delivery time script. Your first file is:

```
delivery_time.py
```

The next day, you improve the logic. You do not want to lose the old version, so you save a copy:

```
delivery_time_V2.py
```

A week later, you have:

```
delivery_time_V2_final.py
delivery_time_V2_final_FIXED.py
delivery_time_V2_final_FIXED_reviewed_by_Sarah.py
```

This system breaks fast. You forget which file is the latest. Sarah edits the wrong copy. Your laptop crashes and you lose three versions at once. Now multiply this problem by 12 engineers and 200 files. The result is chaos.

A **VCS** (Version Control System) solves this problem. A VCS automatically saves the full history of every change in your project. It keeps a single file but tracks every version inside it. You can go back to any past version in seconds. You can see who changed what and when. You can work on the same file as a teammate without overwriting their changes.

**Git** is the most popular VCS in the world. Almost every tech company uses it. So this is the tool we will learn.


## Configuring Git on your computer

### Download and install

Go to the [official Git website](https://git-scm.com/downloads), download the installer for your operating system, and follow the setup instructions.

<img src="https://ai-fullstack-assets.s3.eu-west-3.amazonaws.com/M01-AI_Engineering_Foundation/AIFS-M01-D03-Git_download.png" />

### Configure your username and email

Before you can use Git at FreshCart, you have to tell Git who you are. Every change you save will be linked to your name and email. This is how your team knows who wrote each line of code. You only have to do this configuration once per computer.

Open your terminal and type:

```bash
git config --global user.name
```

This command shows your current Git username. If nothing is set, it returns nothing. To set your username, type:

```bash
git config --global user.name "your_username"
```

For your email, the command works the same way. Use the same email as your GitHub account.

```bash
git config --global user.email "your_email@example.com"
```

<Note type="important">

Use your **professional email** when you work for a company. At FreshCart, you would use your `you@freshcart.com` address. This way, all your work commits stay linked to your work identity, not your personal one.

</Note>


## The core Git commands

Git works with special directories called **repositories**. A repository (or *repo*) is a folder that contains the project's code plus a hidden `.git` subfolder where Git stores the full history.

In your job at FreshCart, you will not create repositories from scratch every day. Most of the time, you will *clone* an existing one from GitHub, work on it, and *push* your changes back. The commands below are the daily toolkit.

### git clone

```bash
git clone https://github.com/freshcart/delivery-time-model.git
```

This downloads the FreshCart `delivery-time-model` repository to your computer. Git copies all the files **and** the full history. You can run this command anywhere on your system. Git creates a new folder with the project name.

### git status

```bash
git status
```

This shows the current state of your repository. You see which files you changed, which new files you created, and which files Git is not tracking yet. *Use this command often.* Run it before every commit and every push. It is your safety check to avoid mistakes.

### git add

```bash
git add delivery_time.py
```

This tells Git: *"I want to save the changes I made in `delivery_time.py` in my next commit."* Only files you explicitly add will be saved. To add all changed files at once, use:

```bash
git add .
```

The dot means "everything in the current directory."

### git commit

```bash
git commit -m "Fix bug in delivery time calculation for warehouse 3"
```

A **commit** is a saved snapshot of your project at one moment in time. The `-m` flag lets you write a short message that explains *what you changed and why*. Good commit messages help your team understand your work months later. Bad messages like `"fix"` or `"update"` are useless when you debug a problem six months later.

### git log

```bash
git log
```

This shows the full history of commits on your branch. You see who made each commit, when, with what message, and a unique commit ID (a long string of letters and numbers). You will use these IDs to go back to a specific version if needed.

### git reset

```bash
git reset commit_identifier
```

This rolls your project back to the version saved at the chosen commit. For example, if a coworker breaks the production code and you need to go back to yesterday's working version, this is the command you use.

### git push

```bash
git push
```

This sends all your local commits to GitHub. Until you push, your work only lives on your laptop. After you push, your code is in the cloud. Your team can see it, review it, and pull it. **If your laptop dies tomorrow, your pushed code is safe.** This alone is a reason to push often. 😊

<Note type="warning" title="One rule that saves you hours">

All Git commands above run inside your *working directory*. Make sure you are inside a Git repository before you run them. Run `pwd` and `git status` first if you are not sure. The only exception is `git clone`. It works anywhere because it creates a new repository.

</Note>


## From Git to GitHub

Git on its own is powerful but limited. It only saves history on **your computer**. If your laptop is stolen, your history is gone. If a coworker wants your code, you have to email it. This does not scale.

<img src="https://ai-fullstack-assets.s3.eu-west-3.amazonaws.com/M01-AI_Engineering_Foundation/AIFS-M01-D03-github_signup.png" />

**GitHub** solves this. GitHub is a cloud platform built on top of Git. You push your local repository to GitHub, and your code lives online in the cloud. Now anyone you authorize can pull it, review it, and contribute. This is how the FreshCart team of 12 engineers works on the same codebase without chaos.

GitHub also adds features that pure Git does not have:

- A web interface to read code in the browser
- Permissions to control who can see and edit each repository
- A code review system through *pull requests*
- Issue tracking for bugs and feature requests
- Project boards to plan work

> *GitHub is to Git what a shared Google Drive is to a single Word document on your laptop.*


### Your personal profile

To use GitHub, go to [github.com](https://github.com/) and create an account with a username, an email, and a password. **Pick your username carefully.** This is the name recruiters will see. Use something close to your real name. Avoid nicknames you would not show in a job interview.

Once your account is ready, your profile shows:

- The repositories you own
- The repositories you star (a star is a public bookmark)
- Your number of followers
- The number of people you follow

To change your settings, click your avatar in the top-right corner of any page. A dropdown menu opens with the *Settings* link.

### Being part of an organization

A real company on GitHub is not a personal profile. It is an **organization**. An organization is a shared GitHub account that hosts a company's repositories. FreshCart, for example, has its own organization at `github.com/freshcart`. Inside this organization:

- All company repositories live in one place
- Admins decide who has access to what
- Teams group engineers by project (the recommendation team, the warehouse team, etc.)

When you join FreshCart, an admin invites you to the FreshCart organization. After you accept, you can see and clone the team's private repositories from your own GitHub account.

> *Your personal profile is your portfolio. Your company organization is your day job. They live side by side on the same GitHub account.*


## Creating a remote repository

Now imagine your manager at FreshCart asks you to start a new internal project: a small Python tool that audits warehouse stock levels every night. You need a fresh repository for this project.

<img src="https://ai-fullstack-assets.s3.eu-west-3.amazonaws.com/M01-AI_Engineering_Foundation/AIFS-M01-D03-github_new_repo.png" />

Go to [github.com/new](https://github.com/new) (or click the green *New* button on your homepage). Then configure the repository step by step.

### Step 1. Choose a clear name

Pick a name that summarizes what is inside. For the audit tool, a good name is `warehouse-stock-audit`. Avoid generic names like `project1`, `test`, or `code`. Six months from now, your future self will thank you.

### Step 2. Choose public or private

You have two access levels:

- **Public**: anyone on the internet can see and clone your code.
- **Private**: only people you invite can see it.

For an internal FreshCart tool, always pick **private**. Internal logic, customer data flows, and warehouse algorithms must never be public. Pick **public** only for your personal portfolio projects, after the code is clean enough to show to a recruiter.

### Step 3. Initialize with a README

Check the box *Initialize this repository with a README*. This creates a `README.md` file at the root of your repository. The README is the front page of your project on GitHub. A good README contains:

- A short description of what the project does
- Installation and usage instructions
- A link to the dataset (if any)
- A list of dependencies and external resources

A README in **Markdown** (a simple text format that lets you write headings, lists, and links) is the first thing a new teammate or a recruiter will read. Treat it as the project's business card.

### Step 4. Add a Python `.gitignore`

In the *Add .gitignore* drop-down menu, select *Python*. This creates a hidden `.gitignore` file that tells Git which files to **never track**. For a Python project, this includes:

- `.ipynb_checkpoints/` (Jupyter backup files)
- `__pycache__/` (Python's compiled cache)
- `.env` (files that contain secrets and API keys)

Without a `.gitignore`, you would push useless files and, worse, possibly leak passwords to GitHub. This is one of the most common mistakes new engineers make.

Once you click *Create repository*, GitHub redirects you to your new repo's homepage. Click the green *Code* button in the top-right corner and copy the URL that ends with `.git`. Then run:

```bash
git clone https://github.com/your_username/warehouse-stock-audit.git
```

You now have a local copy of your new repo. You can start coding inside it.


## Contributing to an open-source project

Many famous Python libraries are **open-source**. This means their code is public, free to use, and improved by contributors from all over the world. You probably already use some of them at FreshCart:

- [pandas](https://github.com/pandas-dev/pandas) for data manipulation
- [scikit-learn](https://github.com/scikit-learn/scikit-learn) for machine learning
- [seaborn](https://github.com/mwaskom/seaborn) for plotting
- [tensorflow](https://github.com/tensorflow/tensorflow) for deep learning

When you become an experienced engineer, you may find a bug in one of these libraries, or have an idea for a new feature. You can fix it yourself and propose your fix back to the project. This is called *contributing*. Contributors do not have direct write access to a famous library. So the workflow uses two extra steps: **fork** and **pull request**.

### Step 1. Fork the repository

A **fork** is your personal copy of someone else's repository, hosted on your GitHub account. You click the *Fork* button on the original repo's page (for example, on `github.com/scikit-learn/scikit-learn`). GitHub creates `github.com/your_username/scikit-learn`. You now own this copy and can do anything inside it.

### Step 2. Clone, change, and push

You use the Git commands you already know:

```bash
git clone https://github.com/your_username/scikit-learn.git
cd scikit-learn
# edit the code, fix the bug
git add .
git commit -m "Fix divide by zero error in train_test_split when n_samples=0"
git push
```

Your fix is now on your fork on GitHub. But the original `scikit-learn` repo does not know about your fix yet.

### Step 3. Send a pull request

A **pull request** (often called a *PR*) is a formal request that says: *"Here is my change. Please review it and pull it into the main project."* You open a pull request from your fork's page on GitHub. The scikit-learn maintainers see your PR. They review your code, ask questions, suggest improvements, and finally accept or reject it.

Adding a contribution to a major open-source library is a strong signal on a CV. It shows recruiters you can read someone else's code, follow a coding standard, and work with a team you do not know personally.


## A simple Git and GitHub team workflow

Almost every tech company today hosts code on GitHub (or a similar platform like GitLab or Bitbucket) and uses Git for version control. This means **you will use these tools on day one of your job**, no matter what your title is.

Each company has its own *Git workflow*: a set of rules about how the team uses branches, who can push where, and how code gets reviewed. Workflows vary, but they all follow the same philosophy. Below is the simplest version, the one you will recognize at FreshCart.

Imagine your teammate Roger is a skilled data scientist who wants to add a new feature to scikit-learn. Here is the full path of his work:

![](https://full-stack-assets.s3.eu-west-3.amazonaws.com/images/01-Git_and_github_basics.png)

The same flow applies inside FreshCart, just without the fork step:

1. **Pull** the latest version of the team's repository to your computer.
2. **Create a branch** for your new feature so the main code stays stable.
3. **Code** your change. Run `git status` often.
4. **Commit** your changes with clear messages.
5. **Push** your branch to GitHub.
6. **Open a pull request** so a teammate can review your code.
7. After approval, your change is **merged** into the main branch and goes to production.

This loop happens dozens of times a day across the FreshCart engineering team. Once you internalize it, you can join almost any modern tech team and be productive within hours.

<Note type="tip" title="Pull before you push">

Before you start coding each morning, run `git pull` to download the latest changes from your teammates. Before you push, run `git pull` again. This prevents painful merge conflicts where two people changed the same line of code in different ways.

</Note>
