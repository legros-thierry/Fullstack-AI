# Markdown and README files 📝

## What you will learn in this course 🧐🧐

The `README.md` file is the first page anyone sees when they open your project on GitHub. A clear README answers three simple questions in 30 seconds: *What does this project do? How do I run it? Who can I ask for help?*

The README is written in **Markdown** (a lightweight text format that turns simple symbols like `#`, `*`, and `[ ]()` into headings, bold text, and links).

By the end of this course, you will be able to:

- Identify the parts of a Markdown file and how GitHub renders them
- Write clear headings, lists, bold text, italics, and inline code
- Insert code blocks with the right language for syntax coloring
- Add links and images that turn a plain text file into a real project page
- Structure a professional README that answers a new teammate's questions in under one minute


<Note type="tip" title="Markdown is everywhere">
Markdown is used in README files, GitHub issues, pull request descriptions, Obsidian, ChatGPT, Slack messages, Notion pages, and Jupyter notebooks. Learning Markdown takes 15 minutes, but the skill pays back every single working day for the rest of your career.
</Note>

## Headings

Headings are the skeleton of your README. They split the page into sections so a reader can scan it in seconds. In Markdown, you create a heading with the hash symbol `#`. The number of hashes sets the level. **One hash is the biggest. Six hashes is the smallest.**

```markdown
# This is a level 1 heading (the project title)
## This is a level 2 heading (a main section)
### This is a level 3 heading (a sub-section)
#### This is a level 4 heading
##### This is a level 5 heading
###### This is the smallest, level 6
```

In a real README, you almost never go below level 3. More than three levels of nesting means the document is too complex to scan. Keep it simple.

Here is what a good heading structure looks like for your `warehouse-stock-audit` project:

```markdown
# Warehouse stock audit

## What the tool does
## Installation
## Usage
## Configuration
## Contact
```

A new teammate sees this table of contents on GitHub and knows exactly where to find what they need. *That is the whole goal of headings.*


## Text styling

Sometimes you want to highlight a specific word or phrase. Markdown gives you three styles for that.

| Style | Syntax | Example |
|-------|--------|---------|
| *Italic* | `*text*` or `_text_` | *text* |
| **Bold** | `**text**` or `__text__` | **text** |
| ***Bold italic*** | `***text***` | ***text*** |

Use **bold** for keywords a reader must not miss, like the name of a required tool or a warning. Use *italic* for nuance, for the title of a book or a paper, or for a small comment. Do not bold every second word. If everything is bold, nothing is bold.

A real README sentence might look like:

```markdown
**Important:** you need *Python 3.10 or higher* to run this tool.
```

This sentence shows the reader at one glance what is critical (the warning) and what is supporting detail (the version).


## Inline code and code blocks

Code in a README must always be visually separated from prose. Markdown gives you two ways to do this.

### Inline code

For a single word like a filename, a command, or a variable name, wrap it in single backticks `` ` ``.

```markdown
Run the script with `python audit.py` from the project root.
```

This renders the command in a monospaced font with a gray background, so the reader knows it is code, not regular text.

### Code blocks

For more than one line, use **three backticks** before and after the block. You can also add the language name right after the opening backticks. GitHub then colors the syntax automatically.

````markdown
```python
import pandas as pd

def load_stock_data(path):
    df = pd.read_csv(path)
    return df
```
````

The result, when GitHub renders it, looks like this:

```python
import pandas as pd

def load_stock_data(path):
    df = pd.read_csv(path)
    return df
```

You can use any language tag: `python`, `bash`, `sql`, `javascript`, `json`, `yaml`, and many more. Always add the language. It costs nothing and it makes your README much easier to read.

For shell commands, use the `bash` tag:

````markdown
```bash
git clone https://github.com/freshcart/warehouse-stock-audit.git
cd warehouse-stock-audit
pip install -r requirements.txt
```
````

<Note type="tip" title="Why language tags matter">

Without a language tag, code is shown in plain gray. With a tag, keywords, strings, and comments are colored. A reader spots a typo or a wrong import in half the time. *This small habit makes your project look professional.*

</Note>


## Links

A README without links is a dead end. Links connect your project to its dataset, its documentation, its dependencies, and its team contacts. The Markdown syntax is simple:

```markdown
[the visible text](the URL)
```

The text in square brackets is what the reader sees. The URL in parentheses is where the click goes. For example:

```markdown
This tool uses the [pandas](https://pandas.pydata.org/) library for data processing.
```

Always write a meaningful visible text. **Avoid `click here` or `this link`.** A clear link text helps screen readers and helps your reader scan the page.

Bad link:

```markdown
For installation instructions, [click here](https://docs.freshcart.io/install).
```

Good link:

```markdown
See the [installation guide on the FreshCart docs](https://docs.freshcart.io/install).
```

You can also link to another file inside your own repository. Use a relative path:

```markdown
See [CONTRIBUTING.md](./CONTRIBUTING.md) for the contribution rules.
```

This is how big open-source projects link from their README to their license, contribution guide, and code of conduct.


## Images

Images make a README come alive. A logo, a screenshot of the dashboard, or a diagram of the data flow tells a story words cannot. The Markdown syntax for an image is almost the same as for a link, but with a `!` in front:

```markdown
![alt text](image URL)
```

The *alt text* is a short description of the image. Screen readers read it out loud, and it shows up if the image fails to load. Always write one.

Example:

```markdown
![FreshCart logo](https://freshcart.io/assets/logo.png)
```

You can also use images stored inside your own repository. Create a folder called `images/` or `assets/` at the root of the repo and reference the file with a relative path:

```markdown
![Architecture diagram](./images/architecture.png)
```

For a real README, a single architecture diagram at the top of the file is often more useful than three paragraphs of explanation.


## Lists

Lists help readers scan a sequence of items. Markdown supports two kinds of lists.

### Unordered lists

Use `-` (a dash) at the start of each line. The order of items does not matter.

```markdown
- Python 3.10 or higher
- pandas 2.0 or higher
- A FreshCart database access token
```

### Ordered lists

Use `1.`, `2.`, `3.` at the start of each line. The order of items matters. Markdown automatically renumbers if you insert or delete items, so do not worry if you write `1. 1. 1.` everywhere.

```markdown
1. Clone the repository
2. Install the dependencies with `pip install -r requirements.txt`
3. Set the environment variable `FRESHCART_TOKEN`
4. Run `python audit.py`
```

You can also nest lists by indenting with two spaces:

```markdown
- Tools you need
  - Git
  - Python 3.10
  - A code editor
- Files you will create
  - `audit.py`
  - `config.yaml`
```

Keep nesting to a maximum of two levels. More than that becomes hard to read.


## Putting it all together: a complete README

Here is what a clean `README.md` for your `warehouse-stock-audit` tool looks like, with everything you learned above. Open this in a Markdown preview and you will see why each piece is there.

````markdown
# Warehouse stock audit

A nightly script that checks FreshCart warehouse stock levels and sends a Slack
alert when any product drops below its safety threshold.

![Architecture diagram](./images/architecture.png)

## What the tool does

- Connects to the FreshCart product database
- Compares current stock against the safety threshold for each product
- Sends a summary message to the `#warehouse-alerts` Slack channel
- Logs the run result to `audit.log`

## Requirements

- Python 3.10 or higher
- A FreshCart database token (ask the data team)
- A Slack webhook URL

## Installation

1. Clone the repository

   ```bash
   git clone https://github.com/freshcart/warehouse-stock-audit.git
   cd warehouse-stock-audit
   ```

2. Install the dependencies

   ```bash
   pip install -r requirements.txt
   ```

3. Create a `.env` file with your secrets

   ```bash
   FRESHCART_TOKEN=your_token_here
   SLACK_WEBHOOK=your_webhook_here
   ```

## Usage

Run the audit manually:

```bash
python audit.py
```

For a scheduled run every night at 2 AM, add this line to your crontab:

```bash
0 2 * * * /usr/bin/python /path/to/audit.py
```

## Contact

Maintained by the **FreshCart data engineering team**.
For questions, ping `@data-eng` on Slack or open an issue.
````

Notice the structure. A reader sees the project name, the one-line description, and the architecture diagram in the first two seconds. They get to a working installation in under a minute. The code blocks have language tags. The bold text highlights the team name. *This is the level of quality you should aim for in every project README.*



<Note type="tip" title="The 30-second rule">

When your README is finished, give it to a teammate who has never seen the project. Ask them: *"After 30 seconds of reading, what does this tool do?"* If they cannot answer in one sentence, your README is not clear enough. Rewrite the top section.

</Note>

## Write documentation everyone can read ♿

Your README will be read by people with very different tools and abilities. Some teammates use a **screen reader** (software that reads the page out loud for blind and low-vision users). Others read on a phone, in a hurry, or in a language that is not their first. Four simple habits make your Markdown work for all of them.

### Rule 1: use a real heading hierarchy

Screen reader users navigate a page by jumping from heading to heading, the same way your eyes scan it. That only works if the hierarchy is real: exactly one `#` title at the top, and no skipped levels.

Bad:

```markdown
# Warehouse stock audit
### Installation
# Usage
```

Good:

```markdown
# Warehouse stock audit
## Installation
## Usage
### Scheduled runs
```

### Rule 2: write descriptive alt text on every image

The alt text is everything a screen reader user gets from your image. It is also what shows up when the image fails to load. An alt text like `image` tells them nothing. Describe what the image shows.

Bad:

```markdown
![image](./images/sales.png)
```

Good:

```markdown
![Chart showing sales doubling in Q3](./images/sales.png)
```

### Rule 3: give every link a meaningful label

Screen readers can list all the links of a page in one flat list, out of context. Ten links that all say `click here` are useless. The link text alone should tell the reader where the click goes.

Bad:

```markdown
To install the tool, [click here](https://docs.freshcart.io/install).
```

Good:

```markdown
See the [installation guide](https://docs.freshcart.io/install).
```

### Rule 4: prefer plain language and short lists

Big dense paragraphs are hard for everyone: screen reader users, non-native English speakers, dyslexic readers, and any teammate scanning your README two minutes before a meeting. Cut long sentences. Turn any sequence of steps into a short list.

Bad:

```markdown
In order to be able to run the tool it is first necessary to ensure that all of
the dependencies have been installed and that the access token has been properly
configured, after which the script may finally be executed.
```

Good:

```markdown
1. Install the dependencies
2. Set your access token
3. Run the script
```

<Note type="important" title="Accessibility is a professional requirement">

Accessible documentation is not a nice-to-have. It is a growing legal requirement for software in many countries, and it is an explicit criterion of your AIA certification (Architecte en Intelligence Artificielle, the French professional certification this bootcamp prepares you for): your project documentation must be usable and accessible to everyone, including people with disabilities.

</Note>
