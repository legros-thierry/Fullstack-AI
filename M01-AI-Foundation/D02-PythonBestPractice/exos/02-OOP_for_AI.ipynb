# Why OOP matters for AI 🤖

You just built the Ford dealership with four classes: `Car`, `Inventory`, `Customer`, and `Dealership`. Each class held some data and offered methods to work with that data.

Soon you will start working with AI tools in Python. The good news is this. You already know most of what you need. AI tools are built with classes, just like your Ford dealership. When you read AI code, you will see the same pattern you built yourself.

This exercise shows you that pattern in four popular Python tools. You do not need to understand AI yet. You only need to spot the classes, the objects, the attributes, and the methods. That is enough to start reading AI code without feeling lost.

<Note type= "definition" title="Quick recap">

Before you go further, remember these four words. They are the only words you need.

A **class** is a blueprint. `Car` is a class.

An **object** is one thing built from the blueprint. `mustang = Car("Mustang", "Red", 2024, 45000)` creates an object.

An **attribute** is a piece of data stored inside an object. `mustang.price` is an attribute.

A **method** is an action the object can do. `mustang.apply_discount(10)` is a method call. You can spot methods because they end with parentheses.

Keep these four words in mind. You will see them again and again in the examples below.

</Note>

## Task 2: What does `fit()` really do?

In the Ford exercise, you wrote a method called `buy_car()`. When Alice called `alice.buy_car(mustang, inventory)`, things changed. Alice's budget became smaller. Her list of owned cars became longer. The inventory lost one car.

In other words, `buy_car()` changed the attributes of the objects. That is what a method does.

The AI method `fit()` does the exact same thing. When you write `house_price_guesser.fit(old_houses, old_prices)`, the object changes inside. Before you call `fit()`, the object knows nothing. After you call `fit()`, the object has learned from the old houses. The learned information is now stored inside the object as attributes.

```python
house_price_guesser = LinearRegression()
# At this point the object is empty

house_price_guesser.fit(old_houses, old_prices)
# Now the object has learned something
# The result is stored inside the object
```

## Task 4: A tool for working with tables

A tool called **Pandas** helps you work with tables of data. Think of a spreadsheet with rows and columns. Pandas lets you open a file, look at the data, and clean it.

```python
import pandas as pd

# Step 1: create an object from a file
sales_table = pd.read_csv("sales.csv")

# Step 2: check some attributes
print(sales_table.shape)
print(sales_table.columns)

# Step 3: call some methods
summary = sales_table.describe()
clean_table = sales_table.dropna()
```

Look at the difference between `sales_table.shape` and `sales_table.describe()`. One of them has parentheses. The other does not. This is the easy way to tell attributes from methods.

## Task 5: The big picture

You just saw three different tools. Scikit-learn helps a computer guess prices. Hugging Face helps a computer write text. Pandas helps you work with tables.

These three tools do very different jobs. But the code pattern is always the same. First, you create an object from a class. Then, you use attributes to read data from the object. Then, you call methods to do actions.

This is why OOP is so important. The tools you will use every day in AI are built with classes. When you understand classes, you can read any AI tool and follow what is happening. You may not know what every method does. But you will know where to look, what to try, and how to ask the right questions.
