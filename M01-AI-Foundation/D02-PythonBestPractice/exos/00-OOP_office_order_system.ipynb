# The OfficeBox order system 📦

You work at **OfficeBox**, a small company that sells office supplies to other businesses. The sales team wants a simple internal tool to track products and customer orders.

Each task below shows one small step. The code is already inside the code cells, so you can run the notebook from top to bottom and study how each part works.


````python
# Your code here
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def describe(self):
        return f"Product: {self.name}, Price: ${self.price}, Stock: {self.stock}"

    def restock(self, amount):
        self.stock += amount

    def sell(self, amount):
        if self.stock >= amount:
            self.stock -= amount
        else:
            raise ValueError("Not enough stock to sell the requested amount")
````

## Task 5: Create an order item

This is the first step toward composition. One object starts to contain another object. 

Create a class called `OrderItem`. It has two attributes: `product` and `quantity`. The `product` value should be a real `Product` object, not just a string. The `quantity` value is a number.

Create a method called `line_total()` that multiplies the product's price by the quantity. For example, if the product is the notebook and the quantity is 3, then `line_total()` should return 13.50.

Use this data to create an order item:

```python
notebook = Product("Notebook", 4.50, 20)
pen_pack = Product("Pen Pack", 6.00, 15)

item_1 = OrderItem(notebook, 3)
item_2 = OrderItem(pen_pack, 2)
```

````python
# Your code here
class OrderItem:
    def __init__(self, product: Product, quantity: int):
        self.product = product
        self.quantity = quantity

    def line_total(self):
        return self.product.price * self.quantity
````

````python
notebook = Product("Notebook", 4.50, 20)
pen_pack = Product("Pen Pack", 6.00, 15)

item_1 = OrderItem(notebook, 3)
item_2 = OrderItem(pen_pack, 2)

print(item_1.line_total())
print(item_2.line_total())
````

````python
class Order:
    def __init__(self, customer: Customer):
        self.customer = customer
        self.items: list[OrderItem] = []

    def add_item(self, product: Product, quantity: int):
        item = OrderItem(product, quantity)
        self.items.append(item)

    def total_price(self):
        total = 0
        for item in self.items:
            total += item.line_total()
        return total
````
