# Ford Manufacturing System 🚗

You work as a junior developer at a Ford dealership in Detroit. The sales team tracks cars, customers, and sales with paper forms and one shared spreadsheet. Mistakes happen every week. A car gets sold twice. A customer's budget gets forgotten. The manager asks you to build a small Python program that models the dealership.

## Task 1: Create the `Car` class

Start with a class called `Car`. A car has a `model`, a `color`, a `year`, and a `price`. The `__init__()` method stores these four values inside each object.

Add a method called `get_info()` that returns one short sentence with the year, color, model, and price. The return value should look like this: `2024 Red Mustang - $45000`.

Add a second method called `apply_discount()`. This method takes a percentage as a parameter. Inside the method, calculate the discount amount first. Then subtract the discount amount from the price. Update the `price` attribute with the new value. At the end, print the new price with a message like `New price after 10% discount: $40500.0`.

Use this data to test your class:

```python
mustang = Car("Mustang", "Red", 2024, 45000)
print(mustang.get_info())
mustang.apply_discount(10)
```

````python
# Your code here
class Car:
    def __init__(self, model: str, color: str, year: int, price: float) -> None:
        self.model = model
        self.color = color
        self.year = year
        self.price = price

    def get_info(self):
        return f"{self.year} {self.color} {self.model} - ${self.price}"

    def apply_discount(self, percentage: int) -> None:
        self.price *= (1 - percentage / 100)
````

````python
# Test the Car class
mustang = Car("Mustang", "Red", 2024, 45000)
print(mustang.get_info())
mustang.apply_discount(10)
print(f"After discount: {mustang.get_info()}")
````

## Task 2: Build the `Inventory` class

The dealership parks many cars on the lot. You need a place to keep track of them all. Create a class called `Inventory` that holds a list of `Car` objects.

The `__init__()` method takes no extra parameters. Inside the constructor, create an empty list called `cars` and store it as an attribute.

Add a method called `add_car()` that takes a `Car` object as a parameter and appends it to the `cars` list. Add a second method called `remove_car()` that takes a `Car` object and removes it from the `cars` list.

The manager wants to know the total value of all cars in stock so she can report it to the finance team. Add a method called `get_total_value()`. Inside the method, start with a variable called `total` and set it to 0. Then loop through each car in `self.cars` and add the car's price to `total`. At the end, return `total`.

Add one more method called `display_inventory()`. This method prints the number of cars in stock first. Then it loops through each car and prints its information on a new line. Use the `get_info()` method from Task 1 to get each car's details.

Use this data to test your class:

```python
inventory = Inventory()

car1 = Car("Mustang", "Red", 2024, 45000)
car2 = Car("F-150", "Blue", 2024, 52000)
car3 = Car("Explorer", "Black", 2024, 47000)

inventory.add_car(car1)
inventory.add_car(car2)
inventory.add_car(car3)

inventory.display_inventory()

total_value = inventory.get_total_value()
print(f"Total inventory value: ${total_value}")
```

You should see three cars listed and a total value near $144,000.

````python
# Your code here
class Inventory:
    def __init__(self):
        self.cars = []

    def add_car(self, car: Car) -> None:
        self.cars.append(car)

    def remove_car(self, car: Car) -> None:
            self.cars.remove(car)

    def get_total_value(self) -> float:
         total = 0
         for car in self.cars:
              total += car.price
         return total

    def display_inventory(self) -> None:
        print(f"il y a {len(self.cars)} voitures dans l'inventory")
        for car in self.cars:
            print(car.get_info())
````

````python
# Test the Inventory class
inventory = Inventory()

car1 = Car("Mustang", "Red", 2024, 45000)
car2 = Car("F-150", "Blue", 2024, 52000)
car3 = Car("Explorer", "Black", 2024, 47000)

inventory.add_car(car1)
inventory.add_car(car2)
inventory.add_car(car3)

inventory.display_inventory()

total_value = inventory.get_total_value()
print(f"Total inventory value: ${total_value}")
````

## Task 3: Build the `Customer` class

A customer walks in with a budget in mind. The sales assistant needs two things. First, check if the customer can afford a specific car. Second, record the purchase if the customer decides to buy.

Create a class called `Customer`. The `__init__()` method takes a `name` and a `budget`. Store both values as attributes. Also create an empty list called `owned_cars` and store it as an attribute. This list tracks every car the customer buys.

Add a method called `can_afford()` that takes a `Car` object as a parameter. Inside the method, compare the budget to the car's price. Return `True` if the budget is greater than or equal to the price. Return `False` if it is not.

Add a second method called `buy_car()`. This method takes two parameters: a `Car` object and an `Inventory` object. The method handles the full purchase step by step.

First, call `can_afford()` to check the budget. If the customer cannot afford the car, print a message that explains the problem and return `False`. If the customer can afford the car, do three things. Remove the car from the inventory using `inventory.remove_car()`. Add the car to the customer's `owned_cars` list. Subtract the car's price from the budget. At the end, print a message that confirms the purchase and shows the remaining budget. Return `True`.

Use this data to test your class:

```python
test_inventory = Inventory()
mustang = Car("Mustang", "Red", 2024, 45000)
f150 = Car("F-150", "Blue", 2024, 52000)
test_inventory.add_car(mustang)
test_inventory.add_car(f150)

alice = Customer("Alice", 50000)
alice.buy_car(mustang, test_inventory)

bob = Customer("Bob", 30000)
bob.buy_car(f150, test_inventory)
```

Alice has enough money to buy the Mustang. Bob does not have enough money for the F-150.

````python
# Your code here
class Customer:
    def __init__(self, name: str, budget: float) -> bool:
        self.name = name
        self.budget = budget
        self.owned_cars = []

    def can_afford(self, car: Car) -> bool:
        return self.budget >= car.price

    def buy_car(self, car: Car, inventory: Inventory) -> bool:
        if not self.can_afford(car):
            print(f"{self.name} : Budget trop limite pour l'achat de cette voiture")
            return False
        else:
            inventory.remove_car(car)
            self.owned_cars.append(car)
            self.budget -= car.price
            print(f"{self.name} : Ok la vente est faite !")
            print(f"Budget client restant: {self.budget}")
            return True
````

````python
# Test the Customer class
test_inventory = Inventory()
mustang = Car("Mustang", "Red", 2024, 45000)
f150 = Car("F-150", "Blue", 2024, 52000)
test_inventory.add_car(mustang)
test_inventory.add_car(f150)

alice = Customer("Alice", 50000)
alice.buy_car(mustang, test_inventory)

bob = Customer("Bob", 30000)
bob.buy_car(f150, test_inventory)
````

## Task 4: Build the `Dealership` class

The last class brings everything together. The `Dealership` class owns one `Inventory` object and keeps track of total revenue and a list of customers. The manager uses it to generate a daily report.

Create a class called `Dealership`. The `__init__()` method takes a `name` and a `location`. Store both values as attributes. Inside the constructor, also create three more attributes. Create a new `Inventory` object and store it as `self.inventory`. Set `self.revenue` to 0. Create an empty list called `customers` and store it as `self.customers`.

Add a method called `register_customer()` that takes a `Customer` object and appends it to the `customers` list.

Add a second method called `sell_car()`. This method takes a `Customer` object and a `Car` object. First, save the car's price in a variable called `sale_price` before anything else happens. Then call the customer's `buy_car()` method with the car and the dealership's inventory. If the sale goes through, add the sale price to `self.revenue` and return `True`. If the sale fails, return `False`.

Add a third method called `generate_report()`. This method prints a short summary with four lines. First, print the dealership name and location. Second, print the number of cars left in stock. Third, print the number of registered customers. Fourth, print the total revenue.

Use this data to test your class:

```python
ford_detroit = Dealership("Ford Detroit", "Detroit, Michigan")
ford_detroit.generate_report()
```

At this point, the report should show 0 cars in stock, 0 customers, and $0 in revenue.

````python
# Your code here
class Dealership:
    def __init__(self, name, location):
        self.name = name
        self.location = location
        self.inventory = Inventory()
        self.revenue = 0
        self.customers = []

    def register_customer(self, customer):
        self.customers.append(customer)

    def sell_car(self, car, customer) -> bool:
        sale_price = car.price
        if customer.buy_car(car, self.inventory):
            self.revenue += sale_price
            return True
        else:
            return False
        
    def generate_report(self) -> str:
        report = f"Dealership: {self.name}\n"
        report += f"Location: {self.location}\n"
        report += f"Revenue: ${self.revenue}\n"
        report += f"Inventory: {len(self.inventory.cars)} cars\n"
        report += f"Customers: {len(self.customers)}\n"
        return report
````

````python
# Test the Dealership class
ford_detroit = Dealership("Ford Detroit", "Detroit, Michigan")
print(ford_detroit.generate_report())
````

## Task 5: Run a daily simulation

Now you put all four classes to work. This task simulates one morning at the Ford Detroit dealership. You open the doors, stock the lot with three cars, register two customers, and try two sales.

Follow these five steps in order.

**Step 1.** Create a `Dealership` object called `ford_detroit` with the name `"Ford Detroit"` and the location `"Detroit, Michigan"`.

**Step 2.** Create three `Car` objects: a red Mustang at $45,000, a blue F-150 at $52,000, and a black Explorer at $47,000. All three cars are from 2024. Add each car to the dealership's inventory. Then call `display_inventory()` to confirm the lot is ready.

**Step 3.** Create two `Customer` objects. Alice has a budget of $50,000. Bob has a budget of $30,000. Register both customers at the dealership.

**Step 4.** Try two sales. Sell the Mustang to Alice. Try to sell the F-150 to Bob. One sale will succeed. One sale will fail because of the budget.

**Step 5.** Call `generate_report()` to print the end-of-day summary.

Read the output from top to bottom. Check that the revenue in the final report matches the price of the car that was sold. Check that the number of cars in stock decreased by one.

````python
# Your code here

# Step 1: open the dealership
ford_detroit = Dealership("Ford Detroit", "Detroit, Michigan")

# Step 2: stock the lot with three cars & add the cars to the inventory
car1 = Car("Mustang", "Red", 2024, 45000)
car2 = Car("F-150", "Blue", 2024, 52000)
car3 = Car("Explorer", "Black", 2024, 47000)
ford_detroit.inventory.add_car(car1)
ford_detroit.inventory.add_car(car2)
ford_detroit.inventory.add_car(car3)

print("=== Morning: lot is ready ===")

# Step 3: register two customers
alice = Customer("Alice", 50000)
bob = Customer("Bob", 30000)
ford_detroit.register_customer(alice)
ford_detroit.register_customer(bob)

# Step 4: try two sales
print("\n=== Sales ===")
ford_detroit.sell_car(car1, alice)
ford_detroit.sell_car(car2, bob)

# Step 5: end-of-day report
print("\n=== End-of-day report ===")
print(ford_detroit.generate_report())
````
