```python
monthly_date = pd.date_range(
    start='2026-01-01', 
    end='2026-12-31', 
    freq='ME'
)

print(f"Voici tous les mois de l'année {monthly_date.year[0]}: {monthly_date.month_name().to_list()}")

for m in monthly_date:
    print(m.strftime("%B %Y"))
```

```python
# Method 2: Random time deltas
base_date = pd.Timestamp('2023-01-01')
random_days = np.random.randint(0, 365, size=100)

# pd.Timedelta creates time differences to add to base date
random_timestamps = [base_date + pd.Timedelta(days=int(days)) for days in random_days]
print(f"Random timestamps sample: {random_timestamps[:5]}")

```
