## Task 4: Add a range slider for daily counts

The yearly histogram hides daily variation. The risk team wants to zoom into specific weeks during major events like the 2011 Tohoku earthquake or the 2010 Haiti earthquake. A **range slider** lets them drag a window across the time axis.

Plotly Express does not support range sliders directly. You switch to **Plotly Graph Objects**, which gives you fine control over every chart element.

### Instructions

In the next code cell, build the same kind of histogram but with daily bins and an interactive range slider:

1. Create an empty figure: `fig = go.Figure()`.
2. Add a histogram trace with `fig.add_trace(...)`. Inside, build the trace with `go.Histogram(...)` and pass:
   - `x=df["Date"]`.
   - `xbins=dict(size="D1")`. The `"D1"` string means "one day per bin". Plotly accepts shortcuts like `"M1"` (month), `"W1"` (week), and `"D1"` (day).
   - `name="Daily events"`.
3. Configure the layout with `fig.update_layout(...)`. Pass:
   - `title="Daily number of events (drag the slider to zoom)"`.
   - `xaxis=dict(title="Date", rangeslider=dict(visible=True), type="date")`. The `rangeslider=dict(visible=True)` line is the one that adds the slider strip below the chart. The `type="date"` setting tells Plotly to treat the axis as time.
   - `yaxis=dict(title="Number of events per day")`.
4. Call `fig.show()`.

<Note type='tip'>
**`go.Figure` workflow.** With Graph Objects you always do the same three steps: create an empty `go.Figure`, add one or more traces with `add_trace`, then configure the look with `update_layout`. Once you internalize that pattern, every Plotly chart is just a variation.
</Note>


````python
# your code here
fig = go.Figure()
fig.add_trace(go.Histogram(x=df["Date"],
                           xbins=dict(size="D1"),
                           name="Daily events"))
fig.update_layout(title="Daily number of events (drag the slider to zoom)",
                  xaxis=dict(title="Date", rangeslider=dict(visible=True), type="date"),
                  yaxis=dict(title="Number of events per day"))
fig.show()
````
