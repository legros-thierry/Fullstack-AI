## Import and export data with Pandas

We already saw that Pandas is a powerful library for data generation. Pandas provides two primary data structures: **Series** and **DataFrame**.

- A **Series** is a one-dimensional labeled array. It can hold any data type. It is similar to a single column in a spreadsheet or a SQL table.
- A **DataFrame** is a two-dimensional table with labeled rows and columns. It can hold mixed types. It is similar to a full spreadsheet or a SQL table.

<Note type="important" title="Organize your files before importing data">

Importing data in Python can be tricky if your files are not well organized or if you do not know exactly where they are stored. Always check or set the correct working directory. Keep your data files in a clear, consistent folder structure. This saves time and prevents frustrating errors when reading data.

For example, you can check and change your working directory like this:

```python
import os

# Check current directory

# Change to the folder where your data is stored
os.chdir("C:/Users/YourName/Data")

# Verify the change
print(os.getcwd())
```

</Note>

#### Importing data from CSV files

```python
import pandas as pd

df = pd.read_csv('data.csv')
print(df.head())
```

The most useful options for `read_csv()` are:

- `sep`: a custom delimiter (the default is comma).
- `header`: which row holds the column names (the default is row 0, the first row).
- `index_col`: which column to use as the index (the default is `None`, so Pandas creates a fresh integer index).
- `parse_dates`: which columns should be parsed as dates (the default is `False`).
- [Full reference here](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_csv.html).

#### Importing data from Excel files

```python
df = pd.read_excel('data.xlsx', sheet_name='Sheet1')
print(df.head())
```

The most useful options for `read_excel()` are:

- `sheet_name`: which sheet to read (the default is 0, the first sheet).
- `header`: which row holds column names (the default is 0).
- `index_col`: which column to use as the index (the default is `None`).
- `usecols`: which columns to keep (the default is `None`, meaning all columns).
- [Full reference here](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_excel.html).

#### Importing data from JSON files

JSON (JavaScript Object Notation) is a lightweight data format. It is easy for humans to read and easy for machines to parse. It is commonly used for **transmitting data in web applications**.

```python

```

The most useful options for `read_json()` are:

- `orient`: the format of the JSON string (the default is `'columns'`).
- `typ`: the type of object to return (the default is `'frame'`).
- `convert_dates`: parse specified columns as dates (the default is `True`).
- [Full reference here](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_json.html).

### Indexing, selecting, and assigning

Pandas selection works through several methods. You select columns by name, rows by position, or specific cells by coordinate. Social media analysis requires precise data selection. You need specific accounts, date ranges, or engagement levels.

````python
# Read the simulated posts data from CSV
try:
    posts_df = pd.read_csv('src/simulated_posts.csv')
except FileNotFoundError:
    posts_df = pd.read_csv(r"/content/sample_data/simulated_posts.csv")

# Select a single column. The result is a Series
post_likes = posts_df['likes']
print("First 5 likes values:")
print(post_likes.head())
````

**Single column.** `posts_df['likes']` returns a one-dimensional Series. Use single brackets and a single string.

**Multiple columns.** `posts_df[engagement_columns]` returns a DataFrame with only the listed columns.

#### Boolean indexing

**Boolean indexing filters data based on conditions.** This is the most common operation in data analysis. You filter by account type, by engagement level, or by time period. Multiple conditions combine with logical operators (`&` for AND, `|` for OR).

````python
# Filter business accounts only
business_mask = posts_df['account_type'] == 'business'
business_posts = posts_df[business_mask]
print(f"Business posts: {len(business_posts)} out of {len(posts_df)}")
````

````python
# Filter posts with high likes
high_likes_mask = posts_df['likes'] > 30
high_likes_posts = posts_df[high_likes_mask]
print(f"Posts with more than 30 likes: {len(high_likes_posts)}")
````

````python
# Combine two conditions: business accounts AND high likes
combined_mask = business_mask & high_likes_mask
top_business_posts = posts_df[combined_mask]
print(f"Top business posts: {len(top_business_posts)}")
````

#### Filtering with .query()

Boolean masks are powerful but can become hard to read with many conditions. The `.query()` method offers a cleaner alternative. You write the filter as a **single string**, using normal column names like variables. Pandas parses the string and applies the filter.

<Note type="tip">
`.query()` is especially useful when you have several conditions or when you want to share readable code with colleagues.
</Note>

````python
# Same filter as before, with .query() instead of boolean masks
high_likes_query = posts_df.query('likes > 30')
print(f"Posts with more than 30 likes: {len(high_likes_query)}")
````

````python
# Combine conditions in a single string
top_business_query = posts_df.query('account_type == "business" and likes > 30')
print(f"Top business posts: {len(top_business_query)}")
````

````python
# Multi-condition filter for the analytics dashboard
viral_videos_query = posts_df.query('content_type == "video" and likes >= 25 and comments >= 5')
print(f"Viral videos: {len(viral_videos_query)}")
````

````python
# You can also reference Python variables with @
min_likes = 30
filtered_by_variable = posts_df.query('likes > @min_likes')
print(f"Posts with likes above the threshold ({min_likes}): {len(filtered_by_variable)}")
````

`posts_df.query('likes > @min_likes')` shows the `@` syntax. The `@` prefix tells Pandas to look up `min_likes` in the surrounding Python code. Without `@`, Pandas would search for a column named `min_likes` and fail.

#### Location-based indexing with loc and iloc

**Location-based indexing uses the `iloc` and `loc` methods.**

- `iloc` selects by **integer position** (0, 1, 2, etc.).
- `loc` selects by **label** (the actual index value or column name).

<Note type="warning" title="iloc vs loc">

`iloc` uses integer positions. `loc` uses index labels. After sorting or filtering, positions and labels are no longer the same. Mixing them up is a common source of off-by-one mistakes.

</Note>

````python
# Select the first 5 rows by position with iloc
first_five_posts = posts_df.iloc[:5]
print(first_five_posts)
````

`posts_df.iloc[:5]` selects rows 0, 1, 2, 3, 4 by **position**. The `5` is **excluded**, just like Python slicing. This always works regardless of the row labels.

````python
# Select specific rows and columns with loc (by label)
selected_columns = ['post_id', 'account_type', 'likes']
post_sample = posts_df.loc[0:4, selected_columns]
print(post_sample)
````

`posts_df.loc[0:4, selected_columns]` selects rows by **label**. With `loc`, the end value `4` is **included**. This catches many beginners. The second argument is a list of column names to keep.

In this dataset, row labels happen to equal positions because the index was not changed. So `iloc[:5]` and `loc[0:4]` return the same rows. But once you filter or sort the DataFrame, labels and positions diverge.

#### The mask method

The `.mask()` method replaces values in a Series or DataFrame **based on a condition**. When the condition is `True`, the value is replaced with whatever you specify (the default is `NaN`, meaning "missing"). It is the opposite of `.where()`.

Use `.mask()` to wipe out values that are too small to be meaningful, or to flag outliers, or to handle suspicious data.

````python
# Look at the first 5 comment counts before any change
print("Comments before mask:")
print(posts_df['comments'].head())

# Replace any comment count of 5 or fewer with 0 (treat them as noise)
low_comments_condition = posts_df['comments'] <= 5
posts_df['comments'] = posts_df['comments'].mask(low_comments_condition, 0)

print("\nComments after mask (values <= 5 replaced with 0):")
print(posts_df['comments'].head())
````

`posts_df['comments'] <= 5` builds a boolean mask. It is `True` for posts with 0, 1, 2, 3, 4, or 5 comments.

`.mask(low_comments_condition, 0)` replaces every `True` position with the value `0`. Values where the condition is `False` are kept unchanged.

Note that `.mask()` returns a **new** Series. We assign it back to `posts_df['comments']` to actually update the DataFrame.

### Data assignment and creating new columns

Assignment modifies existing data. You calculate new metrics, clean messy values, or categorize content. There are several ways to add columns: direct assignment, `.assign()`, `.map()`, and `.apply()`.

<Note type="important" title="In-place modification">

Direct assignment to `posts_df['new_column'] = ...` modifies the DataFrame in place. This saves memory but can also overwrite work by accident. Be intentional when assigning.

</Note>

#### Direct column assignment

The simplest way to create a new column is direct assignment. Build the new column from existing ones using arithmetic. The result is **vectorized** (applied to every row at once, no loop needed).

````python
# Calculate total engagement per post: likes + comments + shares
posts_df['total_engagement'] = posts_df['likes'] + posts_df['comments'] + posts_df['shares']

# clip(lower=1) replaces zeros with 1 to avoid division by zero so that we can calculate an engagement rate even for posts with no likes.
safe_likes = posts_df['likes'].clip(lower=1)

# Calculate engagement_rate: (comments / likes) * 100
posts_df['engagement_rate'] = (posts_df['comments'] / safe_likes) * 100
````

````python
# Preview the new columns
preview_columns = ['post_id', 'likes', 'comments', 'shares', 'total_engagement', 'engagement_rate']
print(posts_df[preview_columns].head())
````

#### The apply method

Sometimes the new column needs **custom logic** that pure arithmetic cannot express. That is the job of `.apply()`. You write a Python function that takes a row (or a value) and returns the result. Pandas calls your function for every row.

`.apply()` is more flexible than direct assignment but slower. Use it when the logic is too complex for vector arithmetic.

````python
# Define a function that flags viral posts based on multiple columns
def flag_viral(row):
    likes = row['likes']
    comments = row['comments']
    if likes >= 40 and comments >= 8:
        return True
    else:
        return False

# Apply the function row by row using axis=1
posts_df['is_viral'] = posts_df.apply(flag_viral, axis=1)

# Count how many viral posts we found
viral_count = posts_df['is_viral'].sum()
````

````python
# find out how many viral posts we found
print(f"Viral posts: {viral_count} out of {len(posts_df)}")

posts_df[posts_df['is_viral']]
````

#### The assign method

With `.assign()`, you can create **multiple new columns in a single operation**. This is useful for adding several calculated metrics at once, especially when chaining operations.

````python
# Add two new columns at once with .assign()
posts_df = posts_df.assign(
    total_reactions=posts_df['likes'] + posts_df['comments'],
    is_popular=posts_df['shares'] > 3
)

preview_columns = ['post_id', 'likes', 'comments', 'shares', 'total_reactions', 'is_popular']
print(posts_df[preview_columns].head())
````

`.assign()` returns a **new DataFrame** with the extra columns added. It does not modify the original. We assign the result back to `posts_df` to keep the changes.

Each keyword argument becomes a new column. The keyword name is the column name. The value is the column data.

### Summary functions

Summary functions reveal data patterns at a glance.

- `.info()` shows column names, non-null counts, and dtypes.
- `.describe()` shows count, mean, standard deviation, min, max, and quartiles for numeric columns.
- `.unique()` lists the distinct values in a column.
- `.value_counts()` counts how many times each value appears.

Run these every time you load a new dataset. They surface problems early.

Now let's see the descriptive statistics.

````python
# Basic summary statistics for engagement metrics
print("Post engagement statistics:")
print(posts_df[['likes', 'comments', 'shares']].describe())
````

`.describe()` prints summary statistics for every numeric column.

- `count`: number of non-null values.
- `mean`: average value.
- `std`: standard deviation (a measure of how spread out the values are).
- `min`, `25%`, `50%`, `75%`, `max`: minimum, quartiles, and maximum.

These six numbers give you a fast overview of the distribution. The `50%` value is the median (middle value). If the mean is far above the median, the data is **skewed** (a few large values pull the mean up).

````python
# Count occurrences of each category
print("Account types distribution:")
print(posts_df['account_type'].value_counts())

print("\nContent types distribution:")
print(posts_df['content_type'].value_counts())

# See unique values in a column
unique_types = posts_df['account_type'].unique()
print(f"\nUnique account types: {unique_types}")
````

`.value_counts()` counts how often each value appears. The result is a Series sorted by count (most common first). This is the fastest way to check the distribution of a categorical column.

`.unique()` returns the distinct values as a NumPy array. Use this when you only need to know which values exist, not how many of each.

### Duplicates and string cleanup

Duplicate rows inflate counts and break joins. Two methods help:

- `.duplicated()` returns a boolean Series flagging rows that repeat an earlier row.
- `.drop_duplicates()` returns a copy without those repeats. Use `subset=['column_name']` when only one column must stay unique.

Text columns often arrive with mixed case, extra spaces, or typos. On any string column, use the `.str` accessor. Common patterns are `.str.lower()`, `.str.strip()`, and `.replace()` to map known mistakes to clean labels.

````python
# Small demo so we do not change the main posts table
dup_demo = pd.DataFrame({
    'transaction_id': [1, 2, 2, 3],
    'amount': [10.0, 20.0, 20.0, 30.0],
})

# duplicated() flags repeated rows
duplicate_flags = dup_demo.duplicated().tolist()
print("duplicated() flags:", duplicate_flags)

# drop_duplicates() removes them
deduplicated = dup_demo.drop_duplicates()
print("Rows after drop_duplicates:", len(deduplicated))

# subset keeps unique values in a single column
unique_by_id = dup_demo.drop_duplicates(subset=['transaction_id'])
print("Unique by transaction_id:")
print(unique_by_id)
````

`dup_demo.duplicated()` returns `[False, False, True, False]`. The third row repeats the second, so it gets flagged.

`dup_demo.drop_duplicates()` removes the flagged rows.

`drop_duplicates(subset=['transaction_id'])` only checks the `transaction_id` column. The first occurrence of each ID is kept and the rest are dropped.

````python
# Messy text -> cleaned (typical pattern for free-text fields)
messy_account_types = pd.Series(['  Personal ', 'BUSINESS', ' creator', 'biz'])

# Lower the case
lowered = messy_account_types.str.lower()
print("After lower():", lowered.tolist())

# Strip whitespace from the edges
stripped = lowered.str.strip()
print("After strip():", stripped.tolist())

# Replace a known typo with the correct value
typo_map = {'biz': 'business'}
cleaned = stripped.replace(typo_map)
print("After replace typo:", cleaned.tolist())
````

### Binning numbers and pulling dates apart

`pd.cut()` splits a numeric column into labeled buckets (also called bins or bands). You pass `bins` as the edge values and `labels` for each bucket. Use this for age bands, follower tiers, or any grouping you will report.

Once a column is real datetime (after `pd.to_datetime()`), the `.dt` accessor exposes parts like `.year`, `.month`, and `.day_name()` for the weekday name.

````python
# Bin the likes column into engagement bands
edges = [-1, 10, 25, 40, 100]
labels = ['low', 'medium', 'high', 'viral']
posts_df['likes_band'] = pd.cut(posts_df['likes'], bins=edges, labels=labels)

print("Likes band sample:")
print(posts_df[['likes', 'likes_band']].head())

print("\nDistribution of likes bands:")
print(posts_df['likes_band'].value_counts())
````

### Map functions

Map functions let you **transform data element by element** in a simple and efficient way. In Pandas, the `.map()` method applies a custom function or a predefined mapping to each value in a Series.

`.map()` is most often used with a **dictionary** that translates values from one set to another. This is perfect for converting categorical labels to numeric codes (a common preprocessing step before machine learning).

````python
# Map account types to numeric codes for analysis
account_type_mapping = {'personal': 1, 'business': 2, 'creator': 3}
posts_df['account_type_numeric'] = posts_df['account_type'].map(account_type_mapping)

preview_columns = ['post_id', 'account_type', 'account_type_numeric']
print(posts_df[preview_columns].head())
````

`posts_df['account_type'].map(account_type_mapping)` replaces every value in the column according to the dictionary. `'personal'` becomes 1, `'business'` becomes 2, `'creator'` becomes 3. Any value not in the dictionary becomes `NaN`.

````python
# Map can also use a Python function for more complex logic
def engagement_tier_from_likes(likes):
    if likes < 10:
        return 'low'
    elif likes < 25:
        return 'medium'
    elif likes < 40:
        return 'high'
    else:
        return 'viral'

posts_df['engagement_tier'] = posts_df['likes'].map(engagement_tier_from_likes)

print("Engagement tier distribution:")
print(posts_df['engagement_tier'].value_counts())
````

When `.map()` receives a function, it calls the function once per value. The function must accept one argument and return one value.

Here, `engagement_tier_from_likes` checks the likes count and returns a tier label. The result becomes the new `engagement_tier` column.

<Note type="tip" title="About map() and apply()">

When you call **`.map()`** on a Series or **`.apply()`** on a DataFrame, Pandas returns a **new** Series or DataFrame, depending on what your function returns. **Nothing in your table changes until you assign that return value back to a column.**

- `.map()` works on one column at a time and returns a new Series.
- `.apply(..., axis=1)` runs your function once per row and returns a Series if each call returns a single value.

</Note>
