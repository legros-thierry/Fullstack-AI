## Difference between Pandas and NumPy

You will often hear people compare **NumPy** and **Pandas**, and at first, the difference can feel unclear. 

You can think of **NumPy** as the foundation. It gives you one powerful tool: the **array**. NumPy runs fast and handles numbers very well. When you need to build algorithms, simulate data, apply linear algebra, or run numerical models, you use NumPy.

<Note type="tip" title="Linear Algebra">

In this course, you will not go deep into linear algebra, but it is useful to understand that linear algebra is a branch of mathematics that deals with **vectors, matrices, and linear transformations**. It is fundamental to many areas of data science, including machine learning, computer graphics, and scientific computing.

More here: [Linear Algebra](https://app.jedha.co/track/linear-algebra)

</Note>

But NumPy does not understand **labels**. It does not understand **columns** or **indexes** the way humans do. It works with data as a raw matrix of numbers. 

That is where **Pandas** comes in. Pandas gives you a way to work with structured data. Structured data means data that comes in rows and columns, where each column has a name and each row has a label. **Pandas builds on top of NumPy**. It gives you tools to name columns, to label rows, to filter data based on those names, and to reshape your data in ways that make sense to your analysis.

Don't worry you will understand the relationship best when you watch a **DataFrame** turn into a NumPy **array** and then back again. That transformation shows you what each library sees. **NumPy sees the numbers. Pandas sees the structure**. You need both views to work with data properly.

### Random number generation

NumPy helps generate random numbers through specific mathematical functions. You can control the randomness with **seeds** and specify ranges, probabilities, and distributions. The same seed produces identical random sequences. This reproducibility enables testing and debugging synthetic datasets.

| Method | Description | Use Cases |
| --- | --- | --- |
| `np.random.random()` | Creates floating-point numbers between 0 and 1. You can specify the quantity. | Generates probabilities, ratios, and normalized data. |
| `np.random.rand()` | Creates floating-point numbers between 0 and 1, allowing you to specify the shape of the output array using dimensions (rows, columns). | Useful for generating matrices or multi-dimensional data. |
| `np.random.randint()` | Creates integer sequences within specified bounds. You control minimum value, maximum value, and quantity. | Generates user IDs, post IDs, and count data. |
| `np.random.uniform()` | Creates floating-point numbers within a custom range [low, high]. Unlike `np.random.random()`, you control both the lower and upper bounds. | Use this when you need realistic amounts with a specific range, such as prices between €1 and €2,500 or scores between 0 and 100. |

Let's see how to use these methods to generate synthetic social media data.

````python
# Import necessary libraries

# Control randomness with seed equal to 42 for reproducibility
np.random.seed(42)

# Generate first sequence of 5 random numbers between 0 and 1
random_numbers_1 = np.random.random(5)
print("First sequence:", random_numbers_1)

# Generate second sequence of 5 random numbers between 0 and 1
random_numbers_2 = np.random.rand(5)
print("Second sequence:", random_numbers_2)

# Generate third sequence of 5 random numbers between 0 and 1
random_numbers_3 = np.random.random(5)
print("Third sequence:", random_numbers_3)
````

````python
# Generate a 2D array of random numbers using np.random.random() with shape (3, 4) 3 rows and 4 columns  # noqa: E501
random_2d_array = np.random.random((3, 4))
print("2D array using np.random.random():\n", random_2d_array)

# Generate a 2D array of random numbers using np.random.rand()
random_2d_array_alt = np.random.rand(3, 4)
print("2D array using np.random.rand():\n", random_2d_array_alt)
````

<Note type="tip" title="Understanding Seeds">
By setting a seed, you ensure that the sequence of random numbers generated is **reproducible**. If you run again this notebook with the same seed, you will get the same random numbers. This is crucial for testing and debugging, as it allows you to have consistent results across different runs of your code.
</Note>

````python
# Generate integer sequences
user_ids = np.random.randint(1, 1001, size=10)  # 10 user IDs between 1-1000
print("User IDs:", user_ids)

# Generate counts with different ranges
post_counts = np.random.randint(0, 101, size=20)  # Post counts 0-100
print("Post counts sample:", post_counts[:10])

# Generate categorical indices
category_indices = np.random.randint(0, 3, size=15)  # Indices for 3 categories
print("Category indices:", category_indices)
````

````python
# np.random.uniform() generates floats within a custom range [low, high]
transaction_amounts = np.random.uniform(1, 2500, size=10)
print("Transaction amounts (€):", np.round(transaction_amounts, 2))
````

````python
# Key difference from np.random.random(): you control both bounds
print("np.random.random(5):", np.round(np.random.random(5), 4))   # always 0 to 1
print("np.random.uniform(1, 100, 5):", np.round(np.random.uniform(1, 100, 5), 2))  # 1 to 100  # noqa: E501
````

### Choice selection with custom probabilities

`np.random.choice()` selects items from predefined options. You can specify selection probabilities to match real-world patterns. This method creates categorical data with **realistic distributions**.

````python
# Simple random choice without probabilities
account_types_simple = np.random.choice(['personal', 'business', 'creator'], size=10)

print("Simple choice:", account_types_simple)
````

To simulate account types on a social media platform, with a distribution of 70% "personal," 20% "business," and 10% "creator" accounts, you can use NumPy's `np.random.choice()` function with custom probabilities.

````python
# Choice with realistic probabilities
account_types_weighted = np.random.choice(
    ['personal', 'business', 'creator'],
    size=1000,
    p=[0.7, 0.2, 0.1]  # 70% personal, 20% business, 10% creator
)

print("Weighted choice sample:", account_types_weighted[:10])

# Analyze the distribution of the weighted choices
# Series.value_counts() counts occurrences of each category
print("Weighted choice distribution:", pd.Series(account_types_weighted).value_counts())
````

Boolean data requires probability specification. `np.random.choice()` with True/False options creates boolean arrays with controlled probability distributions.

````python
# Simulate 1000 verified accounts with 5% chance of being True
verified_accounts = np.random.choice([True, False], size=1000, p=[0.05, 0.95])

# Display the count of verified accounts
print(f"Verified accounts: {verified_accounts.sum()} out of {len(verified_accounts)}")
````

````python
# Generate profile picture presence
has_profile_pic = np.random.choice([True, False], size=1000, p=[0.9, 0.1])

# Display the count of accounts with profile pictures
print(f"Accounts with profile pic: {has_profile_pic.sum()} out of {len(has_profile_pic)}")  # noqa: E501
````

### Array creation functions

NumPy provides advanced array creation functions beyond basic range. `np.arange()` creates float sequences. `np.linspace()` creates evenly spaced values. `np.zeros()` and `np.ones()` create arrays filled with specific values.

````python
# NumPy arange for float sequences with step of 0.5
decimal_sequence = np.arange(0, 10, 0.5)
print(f"Decimal sequence: {decimal_sequence}")
````

````python
# Create an array of 25 values evenly spaced between 0 and 24
time_points = np.linspace(0, 24, 25)  # 25 points from 0 to 24 hours
print(f"Time points: {time_points[:5]} ... {time_points[-5:]}")
````

````python
# Create arrays with 0s
zero_array = np.zeros(10, dtype=int)
print(f"Zero array: {zero_array}")
````

````python
# Create arrays with 1s
ones_array = np.ones(5, dtype=bool)
print(f"Ones array: {ones_array}")
````

````python
# Full array with custom value
default_followers = np.full(100, 500)  # 100 accounts with 500 followers each
print(f"Default followers: {default_followers[:10]}")
````

### Dictionary-based DataFrame creation

Dictionaries provide the most intuitive DataFrame creation method. **Dictionary keys become column names. Dictionary values become column data**. 

Values must have equal length or pandas will raise errors.

````python
# Basic dictionary to DataFrame
social_data = {
    'user_id': [101, 102, 103, 104, 105],
    'username': ['luna99', 'max_power', 'sakura_chan', 'data_wolf', 'queenbee'],
    'followers': [4520, 1875, 9200, 640, 7310]
}

print(social_data)
````

### DateTime generation and manipulation 

Dates and times are among the trickiest data types. Start practicing early so they feel easy later.

Pandas provides dedicated datetime types and tools (e.g., `Timestamp`, `DatetimeIndex`, and `Series` with `datetime64[ns]` dtype) to work with temporal data. With them, you can:

* parse strings into dates,
* extract components (year, month, day, hour),
* perform date arithmetic (shifts, offsets, differences),
* resample and roll up by period,
* handle time zones reliably.

That’s why it’s important to learn how to create and manipulate datetime objects when working with time-based data. This skill is essential for analyzing trends, seasonality, and time-series patterns in your datasets.

#### Date range generation

`pd.date_range()` creates sequences of dates with specified frequency and duration. You control start date, end date, number of periods, and frequency pattern. The frequency parameter accepts various values:

| Frequency Code | Description |
| --- | --- |   
|`'D'` | Daily frequency
|`'W'` | Weekly frequency
|`'M'` | Monthly frequency
|`'h'` | Hourly frequency
|`'T'` | Minutely frequency
|`'S'` | Secondly frequency
|`'L'` | Millisecond frequency
|`'Y'` | Yearly frequency
|`'Q'` | Quarterly frequency

Let's see how to use `pd.date_range()` to generate a sequence of dayly timestamps for a year of social media posts.

````python
# Create a date range for daily timestamps in 2026
daily_dates = pd.date_range(start='2026-01-01', end='2026-12-31', freq='D')

print(f"Daily dates count: {len(daily_dates)}")
print(f"First dates: {daily_dates[:5]}")
print(f"Last dates: {daily_dates[-5:]}")
````

````python
# Date range with specific periods
weekly_dates = pd.date_range(start='2026-01-01', periods=52, freq='W')
print(f"Weekly dates: {len(weekly_dates)} periods")
````

````python
# Hourly timestamps
hourly_times = pd.date_range(start='2026-01-01', periods=24, freq='h')
print(f"Hourly timestamps: {hourly_times[:5]}")
````

#### Random timestamp generation

In pandas, you work with time using `pd.Timestamp`. This tool represents one point in time. 

You do not create timestamps one by one. Instead, you usually start with a base or a pattern. Then you apply random changes to simulate natural variation.



````python
# Method 1: Random selection from date range
# Generate 100 random dates in 2023
all_dates = pd.date_range(start='2023-01-01', end='2024-01-01', freq='D')
random_dates = np.random.choice(all_dates, size=100, replace=True)
print(f"Random dates sample: {random_dates[:5]}")
````

````python
# Method 2: Random time deltas
base_date = pd.Timestamp('2023-01-01')
random_days = np.random.randint(0, 365, size=100)

# pd.Timedelta creates time differences to add to base date
random_timestamps = [base_date + pd.Timedelta(days=int(days)) for days in random_days]
print(f"Random timestamps sample: {random_timestamps[:5]}")
````

## Putting it all together

Let's complete an Instagram-like data generation example. For this example, we will create a synthetic dataset of Instagram posts with the following columns:
* `post_id`: A unique identifier for each post (integer).
* `user_id`: A unique identifier for the user who made the post (integer).
* `account_type`: The type of account (categorical: "personal", "business", "creator").
* `post_date`: The date and time when the post was made (datetime).
* `likes`: The number of likes the post received (integer).
* `comments`: The number of comments the post received (integer).   
* `shares`: The number of times the post was shared (integer).
* `content_type`: The type of content in the post (categorical: "photo", "video", "carousel").

````python
import numpy as np
import pandas as pd

# Set seed for reproducibility
np.random.seed(42)

# Number of records to generate
num_posts = 1000

# Generate post_id and user_id
post_ids = np.arange(1, num_posts + 1)
# 200 unique users
user_ids = np.random.randint(1, 201, size=num_posts)

# Generate account types with specified probabilities
account_types = np.random.choice(['personal', 'business', 'creator'],
                                 size=num_posts, p=[0.7, 0.2, 0.1])

# Generate post dates within the last year
post_dates = pd.to_datetime('2023-01-01') + pd.to_timedelta(
    np.random.randint(0, 365, size=num_posts), unit='D'
)

# Generate likes, comments, and shares using simple random integers
# Business accounts tend to get more likes
likes = np.random.randint(0, 41, size=num_posts) + (account_types == 'business') * 10
# Creator accounts tend to get more comments
comments = np.random.randint(0, 11, size=num_posts) + (account_types == 'creator') * 5
# Personal accounts tend to get slightly more shares
shares = np.random.randint(0, 5, size=num_posts) + (account_types == 'personal') * 1

# Generate content types with specified probabilities
content_types = np.random.choice(['photo', 'video', 'carousel'],
                                 size=num_posts, p=[0.6, 0.3, 0.1])

# Create DataFrame
posts_df = pd.DataFrame({
    'post_id': post_ids,
    'user_id': user_ids,
    'account_type': account_types,
    'post_date': post_dates,
    'likes': likes,
    'comments': comments,
    'shares': shares,
    'content_type': content_types
})

# Display the first few rows of the DataFrame
print(posts_df.head())
````

Let's now export these DataFrames to CSV files for further analysis. To do this, you can use the `to_csv()` method provided by Pandas. This method allows you to save your DataFrame to a CSV file easily.

````python
# Export dataset to CSV files
posts_df.to_csv('simulated_posts.csv', index=False)
````

<Note type="tip">
`index=False` prevents Pandas from writing row indices to the file, keeping it clean.
</Note>
