```python
import pandas as pd

path_to_file = "/Users/thierrylegros/00-JEDHA/Fullstack-AI/PrepWork"
data = pd.read_excel(path_to_file + "/COVID-19.xlsx")
```

```python
data.head()
```

```python
us_mask = data["geoId"] == "US"
us_data = data[us_mask]
us_data.head()
```

```python
us_data["cumulated-cases"] = 0
us_data.head()
```

```python
totalrows = len(us_data)
print(totalrows)
```

```python
totalrows
```

```python
us_data = us_data.reset_index(drop=True)
```

```python
us_data.head()
```

```python
for i in range(totalrows - 2, -1, -1):
    us_data.iloc[i, us_data.columns.get_loc("cumulated-cases")] = (
        us_data.iloc[i, us_data.columns.get_loc("cases")]
        + us_data.iloc[i + 1, us_data.columns.get_loc("cumulated-cases")]
    )

us_data.head()
```

```python
us_data.head()
```

```python
newcases = 0
for i in range(totalrows - 2, -1, -1):
    us_data.at[i, "newcases"] = int(us_data.at[i, "cumulated-cases"]) - int(
        us_data.at[i + 1, "cumulated-cases"]
    )
```

```python
us_data.head()
```

```python
growth_ratio = 0
for i in range(totalrows - 2, -1, -1):
    us_data.loc[i, "growth-ratio"] = (
        us_data.loc[i, "newcases"] / us_data.loc[i + 1, "newcases"]
    )
```

```python
us_data.head()
```

```python
us_data.plot(x="dateRep", y="cumulated-cases")
```

```python
for i in range (totalrows - 2, -1, -1):
    us_data.iloc[i, us_data.columns.get_loc("growth-ratio")] = (
        us_data.iloc[i, us_data.columns.get_loc("newcases")]
        / us_data.iloc[i + 1, us_data.columns.get_loc("newcases")]
    )
```

```python
us_data.head()
```

```python
us_data.plot(x="dateRep", y="growth-ratio", figsize=(10, 6))
```

```python
mask_bigvalueOff = us_data["growth-ratio"] < 8
fr_data_filtered = us_data[mask_bigvalueOff]
fr_data_filtered.plot(x="dateRep", y="growth-ratio", figsize=(10, 6), title="Growth Ratio Over Time")
```
