# Random Forest

## What will you learn in this course? 🧐🧐

One major drawback of Decision Trees is that they tend to **overfit**. They adapt too much to the training data, losing generalization ability.
To counter this, we can use a more powerful ensemble algorithm called the **Random Forest** 🌲.

A Random Forest combines **multiple decision trees** to make better, more stable predictions.
In this course, you’ll learn:

* How to interpret Random Forests
* The principle of Bagging (Bootstrap + Aggregating)
* The role of randomness in improving model performance
* How to measure feature importance

## Random Forest 🌲🌲🌲

As its name suggests, a Random Forest is simply a **collection of decision trees** built on different random subsets of the data.
These trees “vote” together to produce a final prediction. This ensemble approach leads to better accuracy and less overfitting than a single tree.

## The Bagging Principle

Bagging is short for **Bootstrap Aggregating**.

The idea is simple: instead of training one model on one dataset, we train **several models on different random samples** of the same data, and then **aggregate** their predictions.

Let $Y$ be the target variable, linked to the explanatory variables $X_1, \dots, X_p$ by a function $f$ such that
$Y = f(X) + \epsilon$ with $n$ observations.

### Bootstrap

Bootstrapping is a resampling technique used to create new training samples from an existing dataset.
From a dataset of $n$ observations, we create a new sample of size $m$ by **drawing with replacement**, meaning the same observation can be drawn multiple times.

This maintains the **same underlying distribution** while creating slightly different datasets.
Each of these samples can then be used to train a separate model, leading to diverse learners.

### Aggregating

Once we’ve drawn $B$ bootstrap samples ${Z_b}_{b \in [[1,B]]}$, we can train $B$ different models.
We then **aggregate** their predictions:

* **Regression (quantitative $Y$):**
  The final prediction is the **average** of all model predictions:
  $$
  \hat{f}*B(x) = \frac{1}{B}\sum*{b=1}^{B} \hat{f}_{Z_b}(x)
  $$

* **Classification (qualitative $Y$):**
  The final prediction is the **majority vote** among all trees:
  $$
  \hat{f}_B(x) = \text{argmax}*j ; \text{Card}{b \mid \hat{f}*{Z_b}(x) = j}
  $$

This process drastically reduces variance compared to a single tree.
However, the challenge lies in ensuring that these samples are sufficiently **independent**, which is not always trivial when $n$ is small.

## Intuitive Understanding 🤓🤓

The Random Forest algorithm builds upon the idea of bagging multiple trees, but adds a **second layer of randomness**.

There are two sources of randomness:

1. **Random sampling of observations** (bootstrap)
2. **Random selection of features** at each split in each tree

This dual randomness has three key advantages:

* It helps trees become more independent from each other.
* It reduces computation per tree.
* It limits the impact of correlated features.

In practice:

* Each tree is usually grown fully (no pruning), with only a small minimum number of observations per leaf.
* For classification, we typically select √p features at each split.
* For regression, we use about p/3 features.

Each tree has a low bias but high variance, and averaging across many trees reduces that variance leading to a model that’s both accurate and robust.

The downside? Random Forests can be large in size (since all trees must be stored) and are **less interpretable** than a single decision tree.

## Feature Importance 📊📊

Random Forests are not only strong predictors, they also provide useful insights into **which features matter most**.

There are several ways to estimate feature importance:

### 1. Mean Decrease in Accuracy

We can measure how much a variable contributes to model accuracy by **randomly shuffling its values** and observing how much the validation error increases.
The greater the drop in accuracy, the more important the variable.

### 2. Mean Decrease in Gini

At each node in a tree, we measure how much a feature **reduces impurity** (heterogeneity).
Summing these decreases across all splits and weighting by the number of samples gives us the **Gini-based importance** for each variable.

### 3. Overall Variable Importance

Finally, we can define the global variable importance for a Random Forest as the **average importance** of each feature across all trees.

Unlike simpler metrics (like F1 or Chi² scores), Random Forest variable importance can capture **non-linear relationships** between features and the target variable which makes it particularly valuable for complex datasets.
